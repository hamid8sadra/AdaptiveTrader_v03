```python
# Cell_05 create_cycle_folders
import json
from datetime import date, timedelta
from pathlib import Path


def _format_ini_date(d):
    """MT5 .ini wants YYYY.MM.DD format."""
    return d.strftime("%Y.%m.%d")


def _write_ini_from_dict(tester_dict, dest_path, symbol, from_date, to_date):
    """
    Write an MT5-style .ini file from a dict shaped like the JSON example.
    tester_dict must contain "Tester" and "TesterInputs" keys.
    """
    tester = dict(tester_dict["Tester"])
    inputs = dict(tester_dict["TesterInputs"])

    # Override the dynamic fields
    tester["Symbol"] = symbol
    tester["FromDate"] = _format_ini_date(from_date)
    tester["ToDate"] = _format_ini_date(to_date)

    lines = ["[Tester]"]
    for key, value in tester.items():
        lines.append(f"{key}={value}")

    lines.append("[TesterInputs]")
    for key, value in inputs.items():
        lines.append(f"{key}={value}")

    dest_path.write_text("\n".join(lines) + "\n", encoding="utf-8")


def create_cycle_folders(config, set_1, set_2):
    """Create one folder per optimization cycle and return the cycle count.

    For each cycle folder, also write <set_name>.ini files derived from the
    given set definitions (dicts matching the JSON schema).
    """
    if isinstance(config, str):
        config = json.loads(config)

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    lookback = int(config["optimization_lookback_days"])

    if interval <= 0:
        raise ValueError("optimization_interval_days must be positive")
    if lookback < 0:
        raise ValueError("optimization_lookback_days must be non-negative")
    if end < start:
        raise ValueError("end_date must be on or after start_date")

    days = (end - start).days
    cycles = (days + interval - 1) // interval  # ceil division

    folder_name = f"{config['expert']}"
    expert_dir = Path("../data/results") / folder_name
    expert_dir.mkdir(parents=True, exist_ok=True)

    # Explicit (name, dict) pairs — no reliance on a 'name' key inside the dict.
    sets = [
        ("set_1", set_1), # ("EA_NBO_WeightedPricePDF_1", set_1),
        ("set_2", set_2), # ("EA_NBO_WeightedPricePDF_2", set_2),
    ]

    for cycle in range(0, cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        from_date = cycle_start - timedelta(days=lookback)

        cycle_dir = expert_dir / f"{config['symbol']}_{cycle_start.isoformat()}"
        cycle_dir.mkdir(parents=True, exist_ok=True)

        for set_name, set_def in sets:
            _write_ini_from_dict(
                tester_dict=set_def,
                dest_path=cycle_dir / f"{set_name}.ini",
                symbol=config["symbol"],
                from_date=from_date,
                to_date=cycle_start,
            )

    return cycles


cycle_count = create_cycle_folders(
    CONFIG, EA_NBO_WeightedPricePDF_1, EA_NBO_WeightedPricePDF_2
)
print(f"Created {cycle_count} cycle folders.")
```

```python
# Cell_06 run_cycle_backtests
def run_cycle_backtests(config, timeout=3600):
    """
    For each cycle folder:
      1. Run set_1.ini and wait until its MT5 report is generated. ignore its report.
      2. Run set_2.ini and wait until its MT5 report is generated. wait for its report.
      3. Copy the set_2 report to the cycle folder as set_2.xml. Save set_2 optimization results next to its INI.

    The set_1 report is ignored.
    Original INI files are restored after execution.
    Wait until the report exists and its size is stable.
    Run one MT5 optimization and wait for its XML report.
    Remove stale report so we don't pick up the previous run.
    Restore the original INI configuration.
    Allow MT5 to finish its shutdown.
    Build the cycle list (mirrors create_cycle_folders logic).
    """
    import shutil
    import subprocess
    import time
    from datetime import date, timedelta
    from pathlib import Path

    terminal_path = config["terminal_path"]
    mql5_path = Path(config["mql5_path"])
    report_path = Path(config["reprort_path"])

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    # Mirror create_cycle_folders logic to enumerate cycles deterministically.
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _wait_for_stable_file(path, timeout=timeout, poll=1.0, stable_for=2.0):
        """Wait until `path` exists, is non-empty, and its size has been stable."""
        deadline = time.time() + timeout
        last_size = -1
        stable_since = None
        while time.time() < deadline:
            if path.exists():
                try:
                    size = path.stat().st_size
                except OSError:
                    size = -1
                if size > 0:
                    if size == last_size:
                        if stable_since is None:
                            stable_since = time.time()
                        elif time.time() - stable_since >= stable_for:
                            return True
                    else:
                        stable_since = None
                        last_size = size
            time.sleep(poll)
        return False

    def _resolve_report_path(ini_tester):
        """Resolve the report path from the INI's Tester section (relative to MQL5 root or absolute)."""
        report = ini_tester.get("Report", "")
        if not report:
            return None
        rp = Path(report)
        if rp.is_absolute():
            return rp
        # MT5 resolves relative report paths under the terminal's MQL5 root parent,
        # but in practice the configured reprort_path is the safest anchor.
        candidate = report_path / rp.name
        if candidate.exists():
            return candidate
        return mql5_path / rp

    def _run_ini(ini_path):
        """Launch MT5 with the given INI and return the subprocess handle."""
        cmd = [terminal_path, f"/config:{ini_path}"]
        return subprocess.Popen(cmd)

    for cycle_dir in cycle_dirs:
        set_1_ini = cycle_dir / "set_1.ini"
        set_2_ini = cycle_dir / "set_2.ini"

        if not set_1_ini.exists() or not set_2_ini.exists():
            print(f"[skip] Missing INI(s) in {cycle_dir}")
            continue

        # Backup original INIs so we can restore them afterwards.
        set_1_backup = set_1_ini.read_bytes()
        set_2_backup = set_2_ini.read_bytes()

        # Parse the Tester section of set_2 to find its report path.
        try:
            import configparser
            parser = configparser.ConfigParser()
            parser.read(set_2_ini, encoding="utf-8")
            set_2_tester = dict(parser["Tester"]) if parser.has_section("Tester") else {}
        except Exception as exc:
            print(f"[warn] Could not parse {set_2_ini}: {exc}")
            set_2_tester = {}

        set_2_report = _resolve_report_path(set_2_tester) or (report_path / "Report_2.xml")

        try:
            # --- Step 1: set_1 (ignored report) ---
            print(f"[cycle] {cycle_dir.name}: running set_1.ini")
            if set_2_report.exists():
                try:
                    set_2_report.unlink()
                except OSError:
                    pass
            proc1 = _run_ini(set_1_ini)
            try:
                proc1.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] set_1 timed out for {cycle_dir.name}; terminating.")
                proc1.terminate()
                try:
                    proc1.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc1.kill()
            # Give MT5 time to release files / shut down between runs.
            time.sleep(3)

            # --- Step 2: set_2 (we keep this report) ---
            print(f"[cycle] {cycle_dir.name}: running set_2.ini")
            if set_2_report.exists():
                try:
                    set_2_report.unlink()
                except OSError:
                    pass

            proc2 = _run_ini(set_2_ini)
            try:
                proc2.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] set_2 timed out for {cycle_dir.name}; terminating.")
                proc2.terminate()
                try:
                    proc2.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc2.kill()

            # --- Step 3: wait for report, then copy it into the cycle folder ---
            if not _wait_for_stable_file(set_2_report, timeout=timeout):
                print(f"[warn] set_2 report not produced for {cycle_dir.name}: {set_2_report}")
                continue

            dest_report = cycle_dir / "set_2.xml"
            shutil.copy2(set_2_report, dest_report)
            print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                  f"({dest_report.stat().st_size} bytes)")

        finally:
            # Restore original INIs.
            try:
                set_1_ini.write_bytes(set_1_backup)
                set_2_ini.write_bytes(set_2_backup)
            except OSError as exc:
                print(f"[warn] Failed to restore INIs in {cycle_dir}: {exc}")
            # Allow MT5 to finish shutting down before the next cycle.
            time.sleep(3)

    print("All cycles completed.")


run_cycle_backtests(CONFIG)
```

```python
# Cell_07 select_top_configs
def select_top_configs(config):
    """
    For each cycle folder:
      1. Read the "set_2.xml" report and extract the optimization results.
      2. Merge rows sharing the same "Result" value.
         - Evaluation columns keep the first row's value.
         - Condition columns keep the shared value only if all rows agree;
           otherwise 0 (per original spec).
         - All other columns keep the most-repeated value in the group
           (ties broken by first occurrence).
      3. Select the top N configurations by the optimization criterion.
      4. Write them to "set_2_selected.csv", preserving the original column
         order from the report header. The "Pass" column is renumbered 1..N.
    """
    import csv
    import re
    import xml.etree.ElementTree as ET
    from collections import Counter
    from pathlib import Path
    from datetime import date, timedelta

    evaluation_columns = [
        "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]
    condition_columns = [
        "O01", "O02", "O03", "O04", "O05", "O06", "O07",
        "C01", "C02", "C03", "C04", "C05", "C06", "C07",
    ]

    top_n = int(config["top_configs"])
    if top_n <= 0:
        raise ValueError("top_configs must be positive")

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _strip_namespaces(xml_text):
        xml_text = re.sub(r'\sxmlns(:\w+)?="[^"]*"', "", xml_text)
        xml_text = re.sub(r'\s\w+:\w+="[^"]*"', "", xml_text)
        xml_text = re.sub(r"<(/?)\w+:", r"<\1", xml_text)
        xml_text = re.sub(r"<\?mso-application[^?]*\?>", "", xml_text)
        return xml_text

    def _parse_report(xml_path):
        """Return (ordered_column_names, list_of_row_values_lists)."""
        raw = xml_path.read_text(encoding="utf-8", errors="replace")
        raw = _strip_namespaces(raw)
        root = ET.fromstring(raw)

        rows = root.findall(".//Row")
        if not rows:
            rows = [el for el in root.iter() if el.tag.endswith("Row")]
        if not rows:
            raise ValueError(f"No Row elements found in {xml_path}")

        def _row_values(row_el):
            values = []
            for cell in row_el:
                if not cell.tag.endswith("Cell"):
                    continue
                data_el = None
                for child in cell:
                    if child.tag.endswith("Data"):
                        data_el = child
                        break
                if data_el is not None:
                    values.append(data_el.text if data_el.text is not None else "")
                else:
                    values.append(cell.text if cell.text is not None else "")
            return values

        header = _row_values(rows[0])

        result_rows = []
        for row_el in rows[1:]:
            vals = _row_values(row_el)
            if not vals or vals[0] == "":
                continue
            try:
                int(vals[0])
            except (TypeError, ValueError):
                continue
            result_rows.append(vals)

        return header, result_rows

    def _to_float(v):
        try:
            return float(v)
        except (TypeError, ValueError):
            return float("-inf")

    def _most_repeated(values):
        """Most common value; ties broken by first occurrence in the list."""
        if not values:
            return ""
        counts = Counter(values)
        max_count = max(counts.values())
        for v in values:  # preserve first-occurrence order
            if counts[v] == max_count:
                return v
        return values[0]

    def _merge_by_result(header, rows):
        try:
            result_idx = header.index("Result")
        except ValueError:
            raise ValueError("'Result' column not found in report header")

        for col in evaluation_columns:
            if col not in header:
                raise ValueError(f"Missing evaluation column: {col}")

        idx_by_name = {name: i for i, name in enumerate(header)}
        known = set(evaluation_columns) | set(condition_columns) | {"Result", "Pass"}

        groups = {}
        for row in rows:
            if result_idx >= len(row):
                continue
            groups.setdefault(row[result_idx], []).append(row)

        merged = []
        for key, group_rows in groups.items():
            merged_row = {}

            # Evaluation columns: first row's value.
            for col in evaluation_columns:
                i = idx_by_name[col]
                merged_row[col] = (
                    group_rows[0][i] if i < len(group_rows[0]) else ""
                )

            # Condition columns: shared value, else 0.
            for col in condition_columns:
                if col not in idx_by_name:
                    continue
                i = idx_by_name[col]
                values = [r[i] for r in group_rows if i < len(r)]
                uniq = set(values)
                merged_row[col] = uniq.pop() if len(uniq) == 1 else "0"

            # All other columns: most-repeated value in the group.
            for col, i in idx_by_name.items():
                if col in known:
                    continue
                values = [r[i] for r in group_rows if i < len(r)]
                merged_row[col] = _most_repeated(values)

            merged_row["Result"] = key
            merged.append(merged_row)

        return merged

    def _sort_value(merged_row):
        custom = _to_float(merged_row.get("Custom", ""))
        if custom != float("-inf"):
            return custom
        return _to_float(merged_row.get("Profit", ""))

    def _write_selected_csv(merged_rows, header, dest_path):
        with dest_path.open("w", encoding="utf-8", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=header, extrasaction="ignore")
            writer.writeheader()
            for rank, row in enumerate(merged_rows, start=1):
                row_out = dict(row)
                row_out["Pass"] = rank
                writer.writerow(row_out)

    for cycle_dir in cycle_dirs:
        report_path = cycle_dir / "set_2.xml"
        if not report_path.exists():
            print(f"[skip] Missing report in {cycle_dir}: {report_path}")
            continue

        try:
            header, rows = _parse_report(report_path)
        except Exception as exc:
            print(f"[warn] Failed to parse {report_path}: {exc}")
            continue

        try:
            merged = _merge_by_result(header, rows)
        except Exception as exc:
            print(f"[warn] Failed to merge results in {cycle_dir}: {exc}")
            continue

        if not merged:
            print(f"[warn] No usable result rows in {cycle_dir}")
            continue

        merged.sort(key=_sort_value, reverse=True)
        selected = merged[:top_n]

        dest_path = cycle_dir / "selected_configs.csv"
        _write_selected_csv(selected, header, dest_path)
        print(f"[cycle] {cycle_dir.name}: wrote {dest_path.name} "
              f"({len(selected)} of {len(merged)} merged configs)")

    print("Top configuration selection completed.")


select_top_configs(CONFIG)

```
