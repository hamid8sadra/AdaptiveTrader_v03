```python
# Cell_05 create_cycle_folders
import json
from datetime import date, timedelta
from pathlib import Path


def _format_ini_date(d):
    """MT5 .ini wants YYYY.MM.DD format."""
    return d.strftime("%Y.%m.%d")


def _write_ini_from_dict(tester_dict, dest_path, symbol, from_date, to_date):
    """
    Write an MT5-style .ini file from a dict shaped like the JSON example.
    tester_dict must contain "Tester" and "TesterInputs" keys.
    """
    tester = dict(tester_dict["Tester"])
    inputs = dict(tester_dict["TesterInputs"])

    # Override the dynamic fields
    tester["Symbol"] = symbol
    tester["FromDate"] = _format_ini_date(from_date)
    tester["ToDate"] = _format_ini_date(to_date)

    lines = ["[Tester]"]
    for key, value in tester.items():
        lines.append(f"{key}={value}")

    lines.append("[TesterInputs]")
    for key, value in inputs.items():
        lines.append(f"{key}={value}")

    dest_path.write_text("\n".join(lines) + "\n", encoding="utf-8")


def create_cycle_folders(config, set_1, set_2):
    """Create one folder per optimization cycle and return the cycle count.

    For each cycle folder, also write <set_name>.ini files derived from the
    given set definitions (dicts matching the JSON schema).
    """
    if isinstance(config, str):
        config = json.loads(config)

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    lookback = int(config["optimization_lookback_days"])

    if interval <= 0:
        raise ValueError("optimization_interval_days must be positive")
    if lookback < 0:
        raise ValueError("optimization_lookback_days must be non-negative")
    if end < start:
        raise ValueError("end_date must be on or after start_date")

    days = (end - start).days
    cycles = (days + interval - 1) // interval  # ceil division

    folder_name = f"{config['expert']}"
    expert_dir = Path("../data/results") / folder_name
    expert_dir.mkdir(parents=True, exist_ok=True)

    # Explicit (name, dict) pairs — no reliance on a 'name' key inside the dict.
    sets = [
        ("set_1", set_1), # ("EA_NBO_WeightedPricePDF_1", set_1),
        ("set_2", set_2), # ("EA_NBO_WeightedPricePDF_2", set_2),
    ]

    for cycle in range(0, cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        from_date = cycle_start - timedelta(days=lookback)

        cycle_dir = expert_dir / f"{config['symbol']}_{cycle_start.isoformat()}"
        cycle_dir.mkdir(parents=True, exist_ok=True)

        for set_name, set_def in sets:
            _write_ini_from_dict(
                tester_dict=set_def,
                dest_path=cycle_dir / f"{set_name}.ini",
                symbol=config["symbol"],
                from_date=from_date,
                to_date=cycle_start,
            )

    return cycles


cycle_count = create_cycle_folders(
    CONFIG, EA_NBO_WeightedPricePDF_1, EA_NBO_WeightedPricePDF_2
)
print(f"Created {cycle_count} cycle folders.")
```

```python
# Cell_06 run_cycle_backtests
def run_cycle_backtests(config, timeout=3600):
    """
    For each cycle folder:
      1. Run set_1.ini and wait until its MT5 report is generated. ignore its report.
      2. Run set_2.ini and wait until its MT5 report is generated. wait for its report.
      3. Copy the set_2 report to the cycle folder as set_2.xml. Save set_2 optimization results next to its INI.

    The set_1 report is ignored.
    Original INI files are restored after execution.
    Wait until the report exists and its size is stable.
    Run one MT5 optimization and wait for its XML report.
    Remove stale report so we don't pick up the previous run.
    Restore the original INI configuration.
    Allow MT5 to finish its shutdown.
    Build the cycle list (mirrors create_cycle_folders logic).
    """
    import shutil
    import subprocess
    import time
    from datetime import date, timedelta
    from pathlib import Path

    terminal_path = config["terminal_path"]
    mql5_path = Path(config["mql5_path"])
    report_path = Path(config["reprort_path"])

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    # Mirror create_cycle_folders logic to enumerate cycles deterministically.
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _wait_for_stable_file(path, timeout=timeout, poll=1.0, stable_for=2.0):
        """Wait until `path` exists, is non-empty, and its size has been stable."""
        deadline = time.time() + timeout
        last_size = -1
        stable_since = None
        while time.time() < deadline:
            if path.exists():
                try:
                    size = path.stat().st_size
                except OSError:
                    size = -1
                if size > 0:
                    if size == last_size:
                        if stable_since is None:
                            stable_since = time.time()
                        elif time.time() - stable_since >= stable_for:
                            return True
                    else:
                        stable_since = None
                        last_size = size
            time.sleep(poll)
        return False

    def _resolve_report_path(ini_tester):
        """Resolve the report path from the INI's Tester section (relative to MQL5 root or absolute)."""
        report = ini_tester.get("Report", "")
        if not report:
            return None
        rp = Path(report)
        if rp.is_absolute():
            return rp
        # MT5 resolves relative report paths under the terminal's MQL5 root parent,
        # but in practice the configured reprort_path is the safest anchor.
        candidate = report_path / rp.name
        if candidate.exists():
            return candidate
        return mql5_path / rp

    def _run_ini(ini_path):
        """Launch MT5 with the given INI and return the subprocess handle."""
        cmd = [terminal_path, f"/config:{ini_path}"]
        return subprocess.Popen(cmd)

    for cycle_dir in cycle_dirs:
        set_1_ini = cycle_dir / "set_1.ini"
        set_2_ini = cycle_dir / "set_2.ini"

        if not set_1_ini.exists() or not set_2_ini.exists():
            print(f"[skip] Missing INI(s) in {cycle_dir}")
            continue

        # Backup original INIs so we can restore them afterwards.
        set_1_backup = set_1_ini.read_bytes()
        set_2_backup = set_2_ini.read_bytes()

        # Parse the Tester section of set_2 to find its report path.
        try:
            import configparser
            parser = configparser.ConfigParser()
            parser.read(set_2_ini, encoding="utf-8")
            set_2_tester = dict(parser["Tester"]) if parser.has_section("Tester") else {}
        except Exception as exc:
            print(f"[warn] Could not parse {set_2_ini}: {exc}")
            set_2_tester = {}

        set_2_report = _resolve_report_path(set_2_tester) or (report_path / "Report_2.xml")

        try:
            # --- Step 1: set_1 (ignored report) ---
            print(f"[cycle] {cycle_dir.name}: running set_1.ini")
            if set_2_report.exists():
                try:
                    set_2_report.unlink()
                except OSError:
                    pass
            proc1 = _run_ini(set_1_ini)
            try:
                proc1.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] set_1 timed out for {cycle_dir.name}; terminating.")
                proc1.terminate()
                try:
                    proc1.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc1.kill()
            # Give MT5 time to release files / shut down between runs.
            time.sleep(3)

            # --- Step 2: set_2 (we keep this report) ---
            print(f"[cycle] {cycle_dir.name}: running set_2.ini")
            if set_2_report.exists():
                try:
                    set_2_report.unlink()
                except OSError:
                    pass

            proc2 = _run_ini(set_2_ini)
            try:
                proc2.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] set_2 timed out for {cycle_dir.name}; terminating.")
                proc2.terminate()
                try:
                    proc2.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc2.kill()

            # --- Step 3: wait for report, then copy it into the cycle folder ---
            if not _wait_for_stable_file(set_2_report, timeout=timeout):
                print(f"[warn] set_2 report not produced for {cycle_dir.name}: {set_2_report}")
                continue

            dest_report = cycle_dir / "set_2.xml"
            shutil.copy2(set_2_report, dest_report)
            print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                  f"({dest_report.stat().st_size} bytes)")

        finally:
            # Restore original INIs.
            try:
                set_1_ini.write_bytes(set_1_backup)
                set_2_ini.write_bytes(set_2_backup)
            except OSError as exc:
                print(f"[warn] Failed to restore INIs in {cycle_dir}: {exc}")
            # Allow MT5 to finish shutting down before the next cycle.
            time.sleep(3)

    print("All cycles completed.")


run_cycle_backtests(CONFIG)
```

```python
# Cell_07 select_top_configs
def select_top_configs(config):
    """
    For each cycle folder:
      1. Read the "set_2.xml" report and extract the optimization results.
      2. Merge rows sharing the same "Result" value.
         - Evaluation columns keep the first row's value.
         - Condition columns keep the shared value only if all rows agree;
           otherwise 0 (per original spec).
         - All other columns keep the most-repeated value in the group
           (ties broken by first occurrence).
      3. Select the top N configurations by the optimization criterion.
      4. Write them to "set_2_selected.csv", preserving the original column
         order from the report header. The "Pass" column is renumbered 1..N.
    """
    import csv
    import re
    import xml.etree.ElementTree as ET
    from collections import Counter
    from pathlib import Path
    from datetime import date, timedelta

    evaluation_columns = [
        "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]
    condition_columns = [
        "O01", "O02", "O03", "O04", "O05", "O06", "O07",
        "C01", "C02", "C03", "C04", "C05", "C06", "C07",
    ]

    top_n = int(config["top_configs"])
    if top_n <= 0:
        raise ValueError("top_configs must be positive")

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _strip_namespaces(xml_text):
        xml_text = re.sub(r'\sxmlns(:\w+)?="[^"]*"', "", xml_text)
        xml_text = re.sub(r'\s\w+:\w+="[^"]*"', "", xml_text)
        xml_text = re.sub(r"<(/?)\w+:", r"<\1", xml_text)
        xml_text = re.sub(r"<\?mso-application[^?]*\?>", "", xml_text)
        return xml_text

    def _parse_report(xml_path):
        """Return (ordered_column_names, list_of_row_values_lists)."""
        raw = xml_path.read_text(encoding="utf-8", errors="replace")
        raw = _strip_namespaces(raw)
        root = ET.fromstring(raw)

        rows = root.findall(".//Row")
        if not rows:
            rows = [el for el in root.iter() if el.tag.endswith("Row")]
        if not rows:
            raise ValueError(f"No Row elements found in {xml_path}")

        def _row_values(row_el):
            values = []
            for cell in row_el:
                if not cell.tag.endswith("Cell"):
                    continue
                data_el = None
                for child in cell:
                    if child.tag.endswith("Data"):
                        data_el = child
                        break
                if data_el is not None:
                    values.append(data_el.text if data_el.text is not None else "")
                else:
                    values.append(cell.text if cell.text is not None else "")
            return values

        header = _row_values(rows[0])

        result_rows = []
        for row_el in rows[1:]:
            vals = _row_values(row_el)
            if not vals or vals[0] == "":
                continue
            try:
                int(vals[0])
            except (TypeError, ValueError):
                continue
            result_rows.append(vals)

        return header, result_rows

    def _to_float(v):
        try:
            return float(v)
        except (TypeError, ValueError):
            return float("-inf")

    def _most_repeated(values):
        """Most common value; ties broken by first occurrence in the list."""
        if not values:
            return ""
        counts = Counter(values)
        max_count = max(counts.values())
        for v in values:  # preserve first-occurrence order
            if counts[v] == max_count:
                return v
        return values[0]

    def _merge_by_result(header, rows):
        try:
            result_idx = header.index("Result")
        except ValueError:
            raise ValueError("'Result' column not found in report header")

        for col in evaluation_columns:
            if col not in header:
                raise ValueError(f"Missing evaluation column: {col}")

        idx_by_name = {name: i for i, name in enumerate(header)}
        known = set(evaluation_columns) | set(condition_columns) | {"Result", "Pass"}

        groups = {}
        for row in rows:
            if result_idx >= len(row):
                continue
            groups.setdefault(row[result_idx], []).append(row)

        merged = []
        for key, group_rows in groups.items():
            merged_row = {}

            # Evaluation columns: first row's value.
            for col in evaluation_columns:
                i = idx_by_name[col]
                merged_row[col] = (
                    group_rows[0][i] if i < len(group_rows[0]) else ""
                )

            # Condition columns: shared value, else 0.
            for col in condition_columns:
                if col not in idx_by_name:
                    continue
                i = idx_by_name[col]
                values = [r[i] for r in group_rows if i < len(r)]
                uniq = set(values)
                merged_row[col] = uniq.pop() if len(uniq) == 1 else "0"

            # All other columns: most-repeated value in the group.
            for col, i in idx_by_name.items():
                if col in known:
                    continue
                values = [r[i] for r in group_rows if i < len(r)]
                merged_row[col] = _most_repeated(values)

            merged_row["Result"] = key
            merged.append(merged_row)

        return merged

    def _sort_value(merged_row):
        custom = _to_float(merged_row.get("Custom", ""))
        if custom != float("-inf"):
            return custom
        return _to_float(merged_row.get("Profit", ""))

    def _write_selected_csv(merged_rows, header, dest_path):
        with dest_path.open("w", encoding="utf-8", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=header, extrasaction="ignore")
            writer.writeheader()
            for rank, row in enumerate(merged_rows, start=1):
                row_out = dict(row)
                row_out["Pass"] = rank
                writer.writerow(row_out)

    for cycle_dir in cycle_dirs:
        report_path = cycle_dir / "set_2.xml"
        if not report_path.exists():
            print(f"[skip] Missing report in {cycle_dir}: {report_path}")
            continue

        try:
            header, rows = _parse_report(report_path)
        except Exception as exc:
            print(f"[warn] Failed to parse {report_path}: {exc}")
            continue

        try:
            merged = _merge_by_result(header, rows)
        except Exception as exc:
            print(f"[warn] Failed to merge results in {cycle_dir}: {exc}")
            continue

        if not merged:
            print(f"[warn] No usable result rows in {cycle_dir}")
            continue

        merged.sort(key=_sort_value, reverse=True)
        selected = merged[:top_n]

        dest_path = cycle_dir / "selected_configs.csv"
        _write_selected_csv(selected, header, dest_path)
        print(f"[cycle] {cycle_dir.name}: wrote {dest_path.name} "
              f"({len(selected)} of {len(merged)} merged configs)")

    print("Top configuration selection completed.")


select_top_configs(CONFIG)

```

```python
# Cell_08 create_selected_inis
import csv
from datetime import date, timedelta
from pathlib import Path


def create_selected_inis(config):
    """
    For each cycle folder:
      1. Read the "selected_configs.csv".
      2. For each row, extract the parameters_columns columns.
      3. For each row, write a new INI file in the cycle folder named
         "selected_<Pass>.ini" (2-digit Pass) with the parameters filled in
         from the CSV row, using EA_NBO_WeightedPricePDF_3 as template.

    The Tester FromDate/ToDate are set to the cycle's start date and end date
    (cycle_start, cycle_end = cycle_start + interval_days).
    All "FILL_VALUE" placeholders in the TesterInputs section are replaced
    with values from the row.
    """
    evaluation_columns = [
        "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]

    # Load template by name (config["expert"] + "_3").
    template_name = f"{config['expert']}_3"
    template = globals().get(template_name)
    if template is None:
        raise ValueError(f"Template '{template_name}' not found in globals")

    if isinstance(config, str):
        import json
        config = json.loads(config)

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    def _format_ini_date(d):
        return d.strftime("%Y.%m.%d")

    def _fill_value(template_value, row, col_name):
        """Replace a 'FILL_VALUE||...' string with the row's value for col_name."""
        if not isinstance(template_value, str):
            return template_value
        if not template_value.startswith("FILL_VALUE"):
            return template_value
        # Strip the leading FILL_VALUE placeholder and the leading '||'.
        rest = template_value[len("FILL_VALUE"):]
        if rest.startswith("||"):
            rest = rest[2:]
        # Replace only the first segment (the actual value), keep the rest.
        value = row.get(col_name, "")
        return f"{value}||{rest}" if rest else str(value)

    def _write_ini(template_dict, row, dest_path, from_date, to_date):
        tester = dict(template_dict["Tester"])
        inputs = dict(template_dict["TesterInputs"])

        tester["Symbol"] = config["symbol"]
        tester["FromDate"] = _format_ini_date(from_date)
        tester["ToDate"] = _format_ini_date(to_date)

        for key, val in inputs.items():
            inputs[key] = _fill_value(val, row, key)

        lines = ["[Tester]"]
        for k, v in tester.items():
            lines.append(f"{k}={v}")
        lines.append("[TesterInputs]")
        for k, v in inputs.items():
            lines.append(f"{k}={v}")
        dest_path.write_text("\n".join(lines) + "\n", encoding="utf-8")

    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_end = cycle_start + timedelta(days=interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if not cycle_dir.exists():
            print(f"[skip] Missing cycle folder: {cycle_dir}")
            continue

        csv_path = cycle_dir / "selected_configs.csv"
        if not csv_path.exists():
            print(f"[skip] Missing CSV: {csv_path}")
            continue

        with csv_path.open("r", encoding="utf-8", newline="") as f:
            reader = csv.DictReader(f)
            rows = list(reader)

        if not rows:
            print(f"[warn] Empty CSV: {csv_path}")
            continue

        # Validate evaluation columns exist in the CSV header.
        fieldnames = reader.fieldnames or []
        for col in evaluation_columns:
            if col not in fieldnames:
                raise ValueError(f"Missing evaluation column '{col}' in {csv_path}")

        for row in rows:
            try:
                pass_num = int(row["Pass"])
            except (KeyError, ValueError, TypeError):
                print(f"[warn] Bad Pass value in {csv_path}: {row.get('Pass')!r}")
                continue

            dest_path = cycle_dir / f"selected_{pass_num:02d}.ini"
            _write_ini(
                template_dict=template,
                row=row,
                dest_path=dest_path,
                from_date=cycle_start,
                to_date=cycle_end,
            )

        print(f"[cycle] {cycle_dir.name}: wrote {len(rows)} selected INI(s)")

    print("Selected INI generation completed.")


create_selected_inis(CONFIG)
```

```python
# Cell_09 run_selected_backtests
def run_selected_backtests(config, timeout=3600):
    """
    For each cycle folder:
      1. look for all INI file in the cycle folder named "selected_<Pass>.ini" (2-digit Pass) with the parameters filled in
      2. run each INI file and wait for its MT5 report to be generated. wait for its report.
      3. Copy the report to the cycle folder as "selected_<Pass>.htm".

    Build the cycle list (mirrors create_cycle_folders logic).
    """
    import re
    import shutil
    import subprocess
    import time
    import configparser
    from datetime import date, timedelta
    from pathlib import Path

    terminal_path = config["terminal_path"]
    mql5_path = Path(config["mql5_path"])
    report_path = Path(config["reprort_path"])

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _wait_for_stable_file(path, timeout=timeout, poll=1.0, stable_for=2.0):
        """Wait until `path` exists, is non-empty, and its size has been stable."""
        deadline = time.time() + timeout
        last_size = -1
        stable_since = None
        while time.time() < deadline:
            if path.exists():
                try:
                    size = path.stat().st_size
                except OSError:
                    size = -1
                if size > 0:
                    if size == last_size:
                        if stable_since is None:
                            stable_since = time.time()
                        elif time.time() - stable_since >= stable_for:
                            return True
                    else:
                        stable_since = None
                        last_size = size
            time.sleep(poll)
        return False

    def _resolve_report_path(ini_tester):
        """Resolve the report path from the INI's Tester section."""
        report = ini_tester.get("Report", "")
        if not report:
            return None
        rp = Path(report)
        if rp.is_absolute():
            return rp
        candidate = report_path / rp.name
        if candidate.exists():
            return candidate
        return mql5_path / rp

    def _run_ini(ini_path):
        cmd = [terminal_path, f"/config:{ini_path}"]
        return subprocess.Popen(cmd)

    for cycle_dir in cycle_dirs:
        selected_inis = sorted(cycle_dir.glob("selected_*.ini"))
        if not selected_inis:
            print(f"[skip] No selected INIs in {cycle_dir}")
            continue

        for ini_path in selected_inis:
            match = re.fullmatch(r"selected_(\d+)\.ini", ini_path.name)
            if not match:
                continue
            pass_str = match.group(1)
            dest_report = cycle_dir / f"selected_{pass_str}.htm"

            # Parse Tester section to find the report path.
            try:
                parser = configparser.ConfigParser()
                parser.read(ini_path, encoding="utf-8")
                tester = dict(parser["Tester"]) if parser.has_section("Tester") else {}
            except Exception as exc:
                print(f"[warn] Could not parse {ini_path}: {exc}")
                tester = {}

            mt5_report = _resolve_report_path(tester) or (report_path / f"Report_3.htm")

            print(f"[cycle] {cycle_dir.name}: running {ini_path.name}")
            if mt5_report.exists():
                try:
                    mt5_report.unlink()
                except OSError:
                    pass

            proc = _run_ini(ini_path)
            try:
                proc.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] {ini_path.name} timed out for {cycle_dir.name}; terminating.")
                proc.terminate()
                try:
                    proc.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc.kill()

            if not _wait_for_stable_file(mt5_report, timeout=timeout):
                print(f"[warn] Report not produced for {ini_path.name} in {cycle_dir.name}: {mt5_report}")
                time.sleep(3)
                continue

            shutil.copy2(mt5_report, dest_report)
            print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                  f"({dest_report.stat().st_size} bytes)")

            # Allow MT5 to finish its shutdown before the next run.
            time.sleep(3)

    print("All selected backtests completed.")


run_selected_backtests(CONFIG)
```

```python
# Cell_10 Balance_Cycles_Report
"""
Build balance series CSVs for an expert's optimization cycles.

1. create_balance_cycles_csv
   - For each cycle folder, read the selected_<Pass>.htm reports.
   - Write balance_cycles/balance_<cycle_FromDate>.csv with columns: minute, <Pass>.
   - Initial balance = Deposit from <expert>_3["Tester"].
   - minute = minutes since cycle start.
   - Rows = union of all timestamps across the cycle's reports; missing Pass values stay empty.

2. create_balance_all_csv
   - Combine every cycle's balance CSV into balance_all.csv with columns: minute, <cycle_FromDate>.
   - Each cycle's value = sum of its Pass balances at that minute.
   - Rows = union of all timestamps across all cycles; missing cycle values stay empty.

Balances are rounded to 2 decimals only when written to CSV.
"""
import csv
import html
import json
import re
from datetime import date, datetime, timedelta
from html.parser import HTMLParser
from pathlib import Path


class _ReportTableParser(HTMLParser):
    """Collect the rows of every HTML table without requiring pandas/lxml."""
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.tables, self._table, self._row, self._cell = [], None, None, None
    def handle_starttag(self, tag, attrs):
        if tag == 'table': self._table = []
        elif tag == 'tr' and self._table is not None: self._row = []
        elif tag in ('td', 'th') and self._row is not None: self._cell = []
        elif tag == 'br' and self._cell is not None: self._cell.append(' ')
    def handle_data(self, data):
        if self._cell is not None: self._cell.append(data)
    def handle_endtag(self, tag):
        if tag in ('td', 'th') and self._cell is not None:
            self._row.append(' '.join(''.join(self._cell).split()))
            self._cell = None
        elif tag == 'tr' and self._row is not None:
            if self._row: self._table.append(self._row)
            self._row = None
        elif tag == 'table' and self._table is not None:
            if self._table: self.tables.append(self._table)
            self._table = None


def _number(value):
    value = html.unescape(str(value)).replace('\xa0', '').replace(' ', '').replace(',', '')
    value = re.sub(r'[^0-9.+-]', '', value)
    return float(value)


def _minute_key(value):
    value = round(value, 6)
    return int(value) if value.is_integer() else value


def _parse_report_time(value):
    value = ' '.join(str(value).split())
    for fmt in ('%Y.%m.%d %H:%M:%S', '%Y.%m.%d %H:%M', '%Y-%m-%d %H:%M:%S', '%Y-%m-%d %H:%M'):
        try:
            return datetime.strptime(value, fmt)
        except ValueError:
            pass
    return None


def _read_report_html(report_path):
    """Decode MT5 reports, which are commonly UTF-16 with a BOM."""
    raw = report_path.read_bytes()
    if raw.startswith((b'\xff\xfe', b'\xfe\xff')):
        return raw.decode('utf-16')
    return raw.decode('utf-8', errors='replace')


def _balance_events(report_path, cycle_start):
    """Return {elapsed_minute: balance} from the Deals table of an MT5 HTML report."""
    parser = _ReportTableParser()
    parser.feed(_read_report_html(report_path))
    events = {}
    for table in parser.tables:
        for index, header in enumerate(table):
            normalized = [re.sub(r'\s+', ' ', cell).strip().lower() for cell in header]
            if 'time' not in normalized or 'balance' not in normalized:
                continue
            time_index, balance_index = normalized.index('time'), normalized.index('balance')
            for row in table[index + 1:]:
                if len(row) <= max(time_index, balance_index):
                    continue
                timestamp = _parse_report_time(row[time_index])
                if timestamp is None:
                    continue
                try:
                    balance = _number(row[balance_index])
                except (TypeError, ValueError):
                    continue
                minute = _minute_key((timestamp - datetime.combine(cycle_start, datetime.min.time())).total_seconds() / 60)
                if minute >= 0:
                    events[minute] = balance  # Last report row wins at an identical timestamp.
    return events


def _cycle_directories(config):
    start, end = date.fromisoformat(config['start_date']), date.fromisoformat(config['end_date'])
    interval = int(config['optimization_interval_days'])
    root = Path('../data/results') / config['expert']
    for offset in range(0, (end - start).days, interval):
        cycle_start = start + timedelta(days=offset)
        cycle_dir = root / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            yield cycle_start, cycle_dir


def create_balance_cycles_csv(config):
    if isinstance(config, str):
        config = json.loads(config)
    template = globals().get(f"{config['expert']}_3")
    if template is None:
        raise ValueError(f"Template '{config['expert']}_3' was not found")
    deposit = float(template['Tester']['Deposit'])
    output_dir = Path('../data/results') / config['expert'] / 'balance_cycles'
    output_dir.mkdir(parents=True, exist_ok=True)

    for cycle_start, cycle_dir in _cycle_directories(config):
        reports = []
        for report in cycle_dir.glob('selected_*.htm'):
            match = re.fullmatch(r'selected_(\d+)\.htm', report.name)
            if match:
                reports.append((int(match.group(1)), report))
        reports.sort()
        if not reports:
            print(f'[skip] No selected reports in {cycle_dir}')
            continue

        pass_names = [f'{pass_number:02d}' for pass_number, _ in reports]
        events_by_pass = {f'{pass_number:02d}': _balance_events(path, cycle_start) for pass_number, path in reports}
        minutes = {0}
        for events in events_by_pass.values():
            minutes.update(events)
        output_path = output_dir / f'balance_{cycle_start.isoformat()}.csv'
        with output_path.open('w', encoding='utf-8', newline='') as handle:
            writer = csv.DictWriter(handle, fieldnames=['minute', *pass_names])
            writer.writeheader()
            for minute in sorted(minutes):
                row = {'minute': minute}
                for pass_name in pass_names:
                    if minute == 0:
                        row[pass_name] = round(deposit, 2)
                    elif minute in events_by_pass[pass_name]:
                        row[pass_name] = round(events_by_pass[pass_name][minute], 2)
                    else:
                        row[pass_name] = ''
                writer.writerow(row)
        print(f'[cycle] Wrote {output_path}')


def create_balance_all_csv(config):
    if isinstance(config, str):
        config = json.loads(config)
    root = Path('../data/results') / config['expert']
    cycle_dir = root / 'balance_cycles'
    cycle_values, all_minutes = {}, {0}
    for cycle_start, _ in _cycle_directories(config):
        path = cycle_dir / f'balance_{cycle_start.isoformat()}.csv'
        if not path.exists():
            print(f'[skip] Missing cycle balance CSV: {path}')
            continue
        with path.open('r', encoding='utf-8', newline='') as handle:
            reader = csv.DictReader(handle)
            pass_names = [name for name in (reader.fieldnames or []) if name != 'minute']
            current = {name: None for name in pass_names}
            totals = {}
            for row in reader:
                minute = _minute_key(float(row['minute']))
                for name in pass_names:
                    if row.get(name, '').strip() != '':
                        current[name] = _number(row[name])
                if all(value is not None for value in current.values()):
                    totals[minute] = sum(current.values())
        cycle_values[cycle_start.isoformat()] = totals
        all_minutes.update(totals)

    columns = list(cycle_values)
    output_path = root / 'balance_all.csv'
    with output_path.open('w', encoding='utf-8', newline='') as handle:
        writer = csv.DictWriter(handle, fieldnames=['minute', *columns])
        writer.writeheader()
        for minute in sorted(all_minutes):
            row = {'minute': minute}
            for cycle_name in columns:
                value = cycle_values[cycle_name].get(minute, '')
                row[cycle_name] = round(value, 2) if value != '' else ''
            writer.writerow(row)
    print(f'Wrote {output_path}')


create_balance_cycles_csv(CONFIG)
create_balance_all_csv(CONFIG)
```

```python
# Cell_11 Performance_Cycles_Report
""" 
Extracts performance metrics from MT5 HTML reports (selected_<Pass>.htm) and writes two CSVs:

- performance_cycles/performance_<cycle_FromDate>.csv — one row per Pass, with all performance metrics.
- performance_all.csv — one row per cycle, aggregating the per-cycle CSVs using SUM, weighted average, MAX/MIN, max drawdown, duration extremes, and recalculated profit/loss ratios.

"""

import csv
import html
import json
import math
import re
from datetime import date, timedelta
from html.parser import HTMLParser
from pathlib import Path


PERFORMANCE_COLUMNS = [
    "Total Net Profit",
    "Gross Profit",
    "Gross Loss",
    "Profit Factor",
    "Recovery Factor",
    "AHPR",
    "GHPR",
    "Total Trades",
    "Expected Payoff",
    "Sharpe Ratio",
    "LR Correlation",
    "LR Standard Error",
    "Profit Trades (% of total)",
    "Loss Trades (% of total)",
    "Largest profit trade",
    "Largest loss trade",
    "Average profit trade",
    "Average loss trade",
    "Maximum consecutive wins ($)",
    "Maximum consecutive losses ($)",
    "Average consecutive wins",
    "Average consecutive losses",
    "Balance Drawdown Absolute",
    "Balance Drawdown Maximal",
    "Balance Drawdown Relative",
    "Equity Drawdown Absolute",
    "Equity Drawdown Maximal",
    "Equity Drawdown Relative",
    "Correlation (Profits,MFE)",
    "Correlation (Profits,MAE)",
    "Correlation (MFE,MAE)",
    "Minimal position holding time",
    "Maximal position holding time",
    "Average position holding time",
]

SUM_COLUMNS = {
    "Total Net Profit",
    "Gross Profit",
    "Gross Loss",
    "Total Trades",
}

WEIGHTED_AVG_COLUMNS = {
    "Expected Payoff",
    "Average profit trade",
    "Average loss trade",
    "Average position holding time",
    "AHPR",
    "GHPR",
    "Profit Factor",
    "Recovery Factor",
    "Sharpe Ratio",
    "LR Correlation",
    "LR Standard Error",
    "Correlation (Profits,MFE)",
    "Correlation (Profits,MAE)",
    "Correlation (MFE,MAE)",
}

MAX_COLUMNS = {
    "Largest profit trade",
    "Maximum consecutive wins ($)",
    "Balance Drawdown Absolute",
    "Balance Drawdown Maximal",
    "Balance Drawdown Relative",
    "Equity Drawdown Absolute",
    "Equity Drawdown Maximal",
    "Equity Drawdown Relative",
    "Maximal position holding time",
}

MIN_COLUMNS = {
    "Largest loss trade",
    "Maximum consecutive losses ($)",
    "Minimal position holding time",
}

RATIO_COLUMNS = {
    "Profit Trades (% of total)",
    "Loss Trades (% of total)",
}


class _PerformanceTableParser(HTMLParser):
    """Read the text content of each row in all tables of an MT5 HTML report."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.tables = []
        self._table = None
        self._row = None
        self._cell = None

    def handle_starttag(self, tag, attrs):
        if tag == "table":
            self._table = []
        elif tag == "tr" and self._table is not None:
            self._row = []
        elif tag in ("td", "th") and self._row is not None:
            self._cell = []
        elif tag == "br" and self._cell is not None:
            self._cell.append(" ")

    def handle_data(self, data):
        if self._cell is not None:
            self._cell.append(data)

    def handle_endtag(self, tag):
        if tag in ("td", "th") and self._cell is not None:
            self._row.append(" ".join("".join(self._cell).split()))
            self._cell = None
        elif tag == "tr" and self._row is not None:
            if self._row:
                self._table.append(self._row)
            self._row = None
        elif tag == "table" and self._table is not None:
            if self._table:
                self.tables.append(self._table)
            self._table = None


def _normalise_label(value):
    value = html.unescape(str(value)).replace("\xa0", " ")
    value = re.sub(r"\s+", " ", value).strip().rstrip(":")
    return value.casefold()


def _read_report_html(report_path):
    raw = report_path.read_bytes()
    if raw.startswith((b"\xff\xfe", b"\xfe\xff")):
        return raw.decode("utf-16")
    return raw.decode("utf-8", errors="replace")


def _number(value, parenthesized=False):
    """Return a numeric report value; optionally prefer a parenthesized amount."""
    text = html.unescape(str(value)).replace("\xa0", " ").strip()
    groups = re.findall(r"\(([^()]*)\)", text)

    if parenthesized and groups:
        text = groups[-1]

    text = text.replace("%", "").replace(" ", "").replace(",", "")
    match = re.search(r"[-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?", text)
    if not match:
        return None

    try:
        return float(match.group())
    except ValueError:
        return None


def _duration_seconds(value):
    """Convert MT5 holding-time strings such as '1:02:03' to seconds."""
    text = html.unescape(str(value)).strip()

    if re.fullmatch(r"\d+(?::\d+){1,2}", text):
        parts = [float(part) for part in text.split(":")]
        if len(parts) == 2:
            return parts[0] * 60 + parts[1]
        return parts[0] * 3600 + parts[1] * 60 + parts[2]

    return _number(text)


def _metric_value(metric, raw_value):
    if metric in {
        "Profit Trades (% of total)",
        "Loss Trades (% of total)",
    }:
        return _number(raw_value, parenthesized=True)

    if metric in {
        "Maximum consecutive wins ($)",
        "Maximum consecutive losses ($)",
    }:
        return _number(raw_value, parenthesized=True)

    if metric in {
        "Minimal position holding time",
        "Maximal position holding time",
        "Average position holding time",
    }:
        return _duration_seconds(raw_value)

    return _number(raw_value)


def _performance_from_report(report_path):
    """
    Extract requested MT5 performance metrics.

    Holding-time values are stored as seconds so all values written to CSV
    remain numeric and use two decimal places.
    """
    parser = _PerformanceTableParser()
    parser.feed(_read_report_html(report_path))

    expected = {_normalise_label(column): column for column in PERFORMANCE_COLUMNS}
    raw_metrics = {}

    for table in parser.tables:
        for row in table:
            # MT5 summary rows usually consist of label/value pairs.
            for index in range(0, len(row) - 1, 2):
                label = _normalise_label(row[index])
                if label in expected and expected[label] not in raw_metrics:
                    raw_metrics[expected[label]] = row[index + 1]

    metrics = {
        column: _metric_value(column, raw_metrics[column])
        if column in raw_metrics
        else None
        for column in PERFORMANCE_COLUMNS
    }

    # Keep trade counts internally for recalculating the combined percentages.
    profit_count = _number(raw_metrics.get("Profit Trades (% of total)", ""))
    loss_count = _number(raw_metrics.get("Loss Trades (% of total)", ""))

    return metrics, profit_count, loss_count


def _cycle_directories(config):
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    root = Path("../data/results") / config["expert"]

    for offset in range(0, (end - start).days, interval):
        cycle_start = start + timedelta(days=offset)
        cycle_dir = root / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            yield cycle_start, cycle_dir


def _csv_value(value):
    """Write every numeric value with exactly two decimal places."""
    if value is None or not isinstance(value, (int, float)) or not math.isfinite(value):
        return ""
    return f"{float(value):.2f}"


def create_performance_cycles_csv(config):
    if isinstance(config, str):
        config = json.loads(config)

    root = Path("../data/results") / config["expert"]
    output_dir = root / "performance_cycles"
    output_dir.mkdir(parents=True, exist_ok=True)

    for cycle_start, cycle_dir in _cycle_directories(config):
        reports = []
        for report_path in cycle_dir.glob("selected_*.htm"):
            match = re.fullmatch(r"selected_(\d+)\.htm", report_path.name)
            if match:
                reports.append((int(match.group(1)), report_path))
        reports.sort()

        if not reports:
            print(f"[skip] No selected reports in {cycle_dir}")
            continue

        output_path = output_dir / f"performance_{cycle_start.isoformat()}.csv"
        with output_path.open("w", encoding="utf-8", newline="") as handle:
            writer = csv.DictWriter(handle, fieldnames=["Pass", *PERFORMANCE_COLUMNS])
            writer.writeheader()

            for pass_number, report_path in reports:
                try:
                    metrics, _, _ = _performance_from_report(report_path)
                except Exception as exc:
                    print(f"[warn] Could not parse {report_path}: {exc}")
                    continue

                row = {"Pass": f"{pass_number:02d}"}
                row.update({column: _csv_value(metrics[column]) for column in PERFORMANCE_COLUMNS})
                writer.writerow(row)

        print(f"[cycle] Wrote {output_path}")


def _finite_values(rows, column):
    values = []
    for row in rows:
        value = row.get(column)
        if isinstance(value, (int, float)) and math.isfinite(value):
            values.append(float(value))
    return values


def _weighted_average(rows, column):
    weighted_sum = 0.0
    total_weight = 0.0

    for row in rows:
        value = row.get(column)
        weight = row.get("Total Trades")
        if (
            isinstance(value, (int, float))
            and isinstance(weight, (int, float))
            and math.isfinite(value)
            and math.isfinite(weight)
            and weight > 0
        ):
            weighted_sum += value * weight
            total_weight += weight

    if total_weight > 0:
        return weighted_sum / total_weight

    values = _finite_values(rows, column)
    return sum(values) / len(values) if values else None


def create_performance_all_csv(config):
    if isinstance(config, str):
        config = json.loads(config)

    root = Path("../data/results") / config["expert"]
    performance_dir = root / "performance_cycles"
    output_path = root / "performance_all.csv"

    output_rows = []

    for cycle_start, cycle_dir in _cycle_directories(config):
        cycle_csv = performance_dir / f"performance_{cycle_start.isoformat()}.csv"
        if not cycle_csv.exists():
            print(f"[skip] Missing cycle performance CSV: {cycle_csv}")
            continue

        rows = []
        with cycle_csv.open("r", encoding="utf-8", newline="") as handle:
            for csv_row in csv.DictReader(handle):
                row = {"Pass": csv_row.get("Pass", "")}
                for column in PERFORMANCE_COLUMNS:
                    row[column] = _number(csv_row.get(column, ""))
                rows.append(row)

        if not rows:
            print(f"[skip] Empty cycle performance CSV: {cycle_csv}")
            continue

        # Read report trade counts to accurately recalculate winning/losing ratios.
        profit_trades = 0.0
        loss_trades = 0.0
        count_found = False

        for report_path in cycle_dir.glob("selected_*.htm"):
            try:
                _, profit_count, loss_count = _performance_from_report(report_path)
            except Exception as exc:
                print(f"[warn] Could not parse {report_path}: {exc}")
                continue

            if profit_count is not None:
                profit_trades += profit_count
                count_found = True
            if loss_count is not None:
                loss_trades += loss_count
                count_found = True

        summary = {"cycle_FromDate": cycle_start.isoformat()}

        for column in PERFORMANCE_COLUMNS:
            values = _finite_values(rows, column)

            if column in SUM_COLUMNS:
                summary[column] = sum(values) if values else None
            elif column in WEIGHTED_AVG_COLUMNS:
                summary[column] = _weighted_average(rows, column)
            elif column in MAX_COLUMNS:
                summary[column] = max(values) if values else None
            elif column in MIN_COLUMNS:
                summary[column] = min(values) if values else None
            elif column in RATIO_COLUMNS:
                if count_found:
                    total = profit_trades + loss_trades
                    if total > 0:
                        count = profit_trades if column.startswith("Profit") else loss_trades
                        summary[column] = count * 100.0 / total
                    else:
                        summary[column] = None
                else:
                    summary[column] = _weighted_average(rows, column)
            else:
                summary[column] = None

        output_rows.append(summary)

    with output_path.open("w", encoding="utf-8", newline="") as handle:
        writer = csv.DictWriter(
            handle,
            fieldnames=["cycle_FromDate", *PERFORMANCE_COLUMNS],
        )
        writer.writeheader()

        for row in output_rows:
            writer.writerow(
                {
                    "cycle_FromDate": row["cycle_FromDate"],
                    **{
                        column: _csv_value(row[column])
                        for column in PERFORMANCE_COLUMNS
                    },
                }
            )

    print(f"Wrote {output_path}")


create_performance_cycles_csv(CONFIG)
create_performance_all_csv(CONFIG)
```

```python
# Cell_12 Balance PNG Reports
""" 
Balance PNG report generator.

Produces three sets of plots from the per-cycle balance CSVs:
1. balance_cycles/balance_<cycle_start>.png — per-cycle line plot (one line per Pass,
   plus thicker combined and average lines), with the y-axis zeroed at the initial balance.
2. balance_all_Overlay.png — two stacked subplots overlaying every cycle: the top shows
   all cycles in blue with combined/average lines, the bottom shows each cycle in its own
   color with a legend.
3. balance_all_Concatenated.png — a single continuous balance curve over real datetime,
   chaining cycles so each starts at the previous cycle's ending balance.
"""


import csv
import json
from datetime import date, datetime, timedelta
from pathlib import Path

import matplotlib.dates as mdates
import matplotlib.pyplot as plt


def _cell12_number(value):
    return float(str(value).replace(",", "").strip())


def _cell12_minute(value):
    value = round(_cell12_number(value), 6)
    return int(value) if value.is_integer() else value


def _cell12_read_balance_csv(csv_path):
    """Read a balance-cycle CSV and forward-fill each Pass balance."""
    with csv_path.open("r", encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        pass_names = [name for name in (reader.fieldnames or []) if name != "minute"]
        rows = list(reader)

    values_by_pass = {pass_name: {} for pass_name in pass_names}
    current = {pass_name: None for pass_name in pass_names}

    for row in rows:
        minute = _cell12_minute(row["minute"])
        for pass_name in pass_names:
            raw_value = row.get(pass_name, "").strip()
            if raw_value:
                current[pass_name] = _cell12_number(raw_value)

            if current[pass_name] is not None:
                values_by_pass[pass_name][minute] = current[pass_name]

    return pass_names, values_by_pass


def _cell12_cycle_directories(config):
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    root = Path("../data/results") / config["expert"]

    total_days = (end - start).days
    cycle_count = (total_days + interval - 1) // interval

    for cycle in range(cycle_count):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_end = cycle_start + timedelta(days=interval)
        cycle_dir = root / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            yield cycle_start, cycle_end, cycle_dir


def _cell12_expand_series(values_by_name, minutes):
    """Forward-fill values for every name at every supplied minute."""
    expanded = {}

    for name, values in values_by_name.items():
        current = None
        series = []

        for minute in minutes:
            if minute in values:
                current = values[minute]
            series.append(current)

        expanded[name] = series

    return expanded


def _cell12_plot_cycle_balance(config):
    root = Path("../data/results") / config["expert"]
    balance_dir = root / "balance_cycles"

    for cycle_start, cycle_end, _ in _cell12_cycle_directories(config):
        csv_path = balance_dir / f"balance_{cycle_start.isoformat()}.csv"
        if not csv_path.exists():
            print(f"[skip] Missing balance CSV: {csv_path}")
            continue

        pass_names, values_by_pass = _cell12_read_balance_csv(csv_path)
        if not pass_names:
            print(f"[skip] No Pass columns in {csv_path}")
            continue

        cycle_minutes = int((cycle_end - cycle_start).total_seconds() / 60)
        minutes = sorted({0, cycle_minutes, *{
            minute
            for values in values_by_pass.values()
            for minute in values
            if 0 <= minute <= cycle_minutes
        }})

        expanded = _cell12_expand_series(values_by_pass, minutes)
        initial_balance = next(
            (
                values[0]
                for values in expanded.values()
                if values and values[0] is not None
            ),
            None,
        )

        if initial_balance is None:
            print(f"[skip] No usable balance data in {csv_path}")
            continue

        fig, ax = plt.subplots(figsize=(15, 8))

        shifted_series = []
        for pass_name, values in expanded.items():
            shifted = [
                value - initial_balance if value is not None else None
                for value in values
            ]
            shifted_series.append(shifted)
            ax.plot(
                minutes,
                shifted,
                color="tab:blue",
                alpha=0.7,
                linewidth=1.4,
                label=f"Pass {pass_name}",
            )

        combined = []
        average = []
        for index in range(len(minutes)):
            available = [
                series[index]
                for series in shifted_series
                if series[index] is not None
            ]
            combined.append(sum(available) if available else None)
            average.append(sum(available) / len(available) if available else None)

        ax.plot(
            minutes,
            combined,
            color="tab:red",
            alpha=0.7,
            linewidth=2.6,
            label="Combined balance",
        )
        ax.plot(
            minutes,
            average,
            color="tab:green",
            alpha=0.7,
            linewidth=2.6,
            label="Average balance",
        )

        ax.set_xlim(0, cycle_minutes)
        ax.set_xlabel("Minutes since cycle start")
        ax.set_ylabel("Balance change")
        ax.set_title(f"Balance by selected configuration — {cycle_start.isoformat()}")
        ax.grid(True, alpha=0.3)
        ax.legend(loc="best", ncol=2)

        ax.text(
            0.01,
            0.01,
            (
                f"Initial balance per Pass: {initial_balance:.2f}\n"
                "Y-axis is shifted so each Pass starts at 0.00.\n"
                "Combined and average balances are calculated across all selected Passes."
            ),
            transform=ax.transAxes,
            fontsize=9,
            verticalalignment="bottom",
            bbox={"boxstyle": "round", "facecolor": "white", "alpha": 0.8},
        )

        output_path = balance_dir / f"balance_{cycle_start.isoformat()}.png"
        fig.tight_layout()
        fig.savefig(output_path, dpi=160, bbox_inches="tight")
        plt.close(fig)
        print(f"[cycle] Wrote {output_path}")


def _cell12_plot_all_overlay(config):
    root = Path("../data/results") / config["expert"]
    balance_dir = root / "balance_cycles"
    cycle_data = {}

    for cycle_start, cycle_end, _ in _cell12_cycle_directories(config):
        csv_path = balance_dir / f"balance_{cycle_start.isoformat()}.csv"
        if not csv_path.exists():
            continue

        pass_names, values_by_pass = _cell12_read_balance_csv(csv_path)
        if not pass_names:
            continue

        cycle_minutes = int((cycle_end - cycle_start).total_seconds() / 60)
        cycle_data[cycle_start.isoformat()] = {
            "cycle_minutes": cycle_minutes,
            "pass_names": pass_names,
            "values_by_pass": values_by_pass,
        }

    if not cycle_data:
        print("[skip] No cycle balance CSVs available for overlay plot")
        return

    longest_cycle_minutes = max(item["cycle_minutes"] for item in cycle_data.values())
    all_minutes = sorted(
        {0, longest_cycle_minutes, *{
            minute
            for item in cycle_data.values()
            for values in item["values_by_pass"].values()
            for minute in values
            if 0 <= minute <= longest_cycle_minutes
        }}
    )

    shifted_cycles = {}
    initial_balances = {}

    for cycle_name, item in cycle_data.items():
        expanded = _cell12_expand_series(item["values_by_pass"], all_minutes)
        initial_balance = next(
            (
                values[0]
                for values in expanded.values()
                if values and values[0] is not None
            ),
            None,
        )
        if initial_balance is None:
            continue

        initial_balances[cycle_name] = initial_balance
        cycle_average = []

        for index in range(len(all_minutes)):
            balances = [
                values[index]
                for values in expanded.values()
                if values[index] is not None
            ]
            cycle_average.append(
                (sum(balances) / len(balances)) - initial_balance
                if balances
                else None
            )

        shifted_cycles[cycle_name] = cycle_average

    if not shifted_cycles:
        print("[skip] No usable cycle balance data for overlay plot")
        return

    combined = []
    average = []

    for index in range(len(all_minutes)):
        values = [
            series[index]
            for series in shifted_cycles.values()
            if series[index] is not None
        ]
        combined.append(sum(values) if values else None)
        average.append(sum(values) / len(values) if values else None)

    fig, (ax1, ax2) = plt.subplots(
        2,
        1,
        figsize=(16, 12),
        sharex=True,
        sharey=True,
    )

    for series in shifted_cycles.values():
        ax1.plot(
            all_minutes,
            series,
            color="tab:blue",
            alpha=0.7,
            linewidth=1.4,
        )

    ax1.plot(
        all_minutes,
        combined,
        color="tab:red",
        alpha=0.7,
        linewidth=2.6,
        label="Combined balance",
    )
    ax1.plot(
        all_minutes,
        average,
        color="tab:green",
        alpha=0.7,
        linewidth=2.6,
        label="Average balance",
    )
    ax1.set_title("All cycles — combined and average balance changes")
    ax1.set_ylabel("Balance change")
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc="best")
    ax1.text(
        0.01,
        0.01,
        (
            "Y-axis is shifted so every cycle starts at 0.00.\n"
            "Combined and average balances are calculated across all cycles."
        ),
        transform=ax1.transAxes,
        fontsize=9,
        verticalalignment="bottom",
        bbox={"boxstyle": "round", "facecolor": "white", "alpha": 0.8},
    )

    palette = [
        "tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple",
        "tab:brown", "tab:pink", "tab:gray", "tab:olive", "tab:cyan",
    ]
    for index, (cycle_name, series) in enumerate(shifted_cycles.items()):
        ax2.plot(
            all_minutes,
            series,
            color=palette[index % len(palette)],
            alpha=0.7,
            linewidth=1.5,
            label=cycle_name,
        )

    ax2.set_title("All cycles — individual cycle balance changes")
    ax2.set_xlabel("Minutes since cycle start")
    ax2.set_ylabel("Balance change")
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc="best", ncol=2, fontsize=8)

    for axis in (ax1, ax2):
        axis.set_xlim(0, longest_cycle_minutes)

    unique_initials = sorted(set(initial_balances.values()))
    initial_text = ", ".join(f"{value:.2f}" for value in unique_initials)
    fig.text(
        0.5,
        0.01,
        f"Initial balance per selected Pass: {initial_text}",
        ha="center",
        fontsize=9,
    )

    output_path = root / "balance_all_Overlay.png"
    fig.tight_layout(rect=(0, 0.03, 1, 1))
    fig.savefig(output_path, dpi=160, bbox_inches="tight")
    plt.close(fig)
    print(f"Wrote {output_path}")


def _cell12_plot_concatenated(config):
    root = Path("../data/results") / config["expert"]
    balance_dir = root / "balance_cycles"
    template = globals().get(f"{config['expert']}_3")

    if template is None:
        raise ValueError(f"Template '{config['expert']}_3' was not found")

    initial_balance = float(template["Tester"]["Deposit"])
    timeline_dates = []
    timeline_balances = []
    running_initial = initial_balance

    for cycle_start, cycle_end, _ in _cell12_cycle_directories(config):
        csv_path = balance_dir / f"balance_{cycle_start.isoformat()}.csv"
        if not csv_path.exists():
            print(f"[skip] Missing balance CSV: {csv_path}")
            continue

        pass_names, values_by_pass = _cell12_read_balance_csv(csv_path)
        if not pass_names:
            continue

        cycle_minutes = int((cycle_end - cycle_start).total_seconds() / 60)
        minutes = sorted({0, cycle_minutes, *{
            minute
            for values in values_by_pass.values()
            for minute in values
            if 0 <= minute <= cycle_minutes
        }})

        expanded = _cell12_expand_series(values_by_pass, minutes)
        cycle_initial = next(
            (
                values[0]
                for values in expanded.values()
                if values and values[0] is not None
            ),
            None,
        )

        if cycle_initial is None:
            continue

        adjusted_values = []
        for index, minute in enumerate(minutes):
            balances = [
                values[index]
                for values in expanded.values()
                if values[index] is not None
            ]
            if not balances:
                continue

            average_balance = sum(balances) / len(balances)
            adjusted_balance = running_initial + (average_balance - cycle_initial)

            timestamp = datetime.combine(cycle_start, datetime.min.time()) + timedelta(
                minutes=minute
            )

            if timeline_dates and timestamp == timeline_dates[-1]:
                timeline_balances[-1] = adjusted_balance
            else:
                timeline_dates.append(timestamp)
                timeline_balances.append(adjusted_balance)

        if timeline_balances:
            running_initial = timeline_balances[-1]

    if not timeline_dates:
        print("[skip] No usable balance data for concatenated plot")
        return

    fig, ax = plt.subplots(figsize=(16, 8))
    ax.plot(
        timeline_dates,
        timeline_balances,
        color="tab:blue",
        alpha=0.7,
        linewidth=2.0,
        label="Concatenated average balance",
    )

    ax.set_title("Concatenated balance across all optimization cycles")
    ax.set_xlabel("Date and time")
    ax.set_ylabel("Balance")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="best")
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m-%d"))
    fig.autofmt_xdate()

    ax.text(
        0.01,
        0.01,
        (
            f"Initial balance for first cycle: {initial_balance:.2f}\n"
            "Each following cycle starts at the final balance of the previous cycle.\n"
            "Cycle balance is the average across its selected Passes."
        ),
        transform=ax.transAxes,
        fontsize=9,
        verticalalignment="bottom",
        bbox={"boxstyle": "round", "facecolor": "white", "alpha": 0.8},
    )

    output_path = root / "balance_all_Concatenated.png"
    fig.tight_layout()
    fig.savefig(output_path, dpi=160, bbox_inches="tight")
    plt.close(fig)
    print(f"Wrote {output_path}")


def create_balance_png_reports(config):
    if isinstance(config, str):
        config = json.loads(config)

    _cell12_plot_cycle_balance(config)
    _cell12_plot_all_overlay(config)
    _cell12_plot_concatenated(config)


create_balance_png_reports(CONFIG)
```

```python
# Cell_13 Performance PNG Reports

"""
Create long, one-column bar-chart reports from the performance CSV files.

Outputs:
- performance_cycles/performance_<cycle_FromDate>.png
- performance_all.png

Every metric is shown in its own subplot row. Numeric labels rendered on bars
always use two decimal places.
"""

import csv
import json
import math
from datetime import date, timedelta
from pathlib import Path

import matplotlib.pyplot as plt


def _cell13_number(value):
    """Convert a CSV value to a finite float, or return None when unavailable."""
    try:
        number = float(str(value).replace(",", "").strip())
    except (TypeError, ValueError):
        return None
    return number if math.isfinite(number) else None


def _cell13_cycle_dates(config):
    """Yield each configured cycle start date."""
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])

    total_days = (end - start).days
    cycle_count = (total_days + interval - 1) // interval

    for cycle in range(cycle_count):
        yield start + timedelta(days=cycle * interval)


def _cell13_read_performance_csv(csv_path, label_column):
    """Read a performance CSV while preserving its source-column order."""
    with csv_path.open("r", encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        rows = list(reader)
        fieldnames = reader.fieldnames or []

    metrics = [column for column in fieldnames if column != label_column]
    labels = [str(row.get(label_column, "")).strip() for row in rows]

    values_by_metric = {}
    for metric in metrics:
        values_by_metric[metric] = [
            _cell13_number(row.get(metric, ""))
            for row in rows
        ]

    return labels, metrics, values_by_metric


def _cell13_add_bar_labels(axis, bars, values):
    """Add two-decimal labels above or below finite bars."""
    finite_values = [value for value in values if value is not None]
    if not finite_values:
        return

    value_range = max(finite_values) - min(finite_values)
    offset = max(abs(max(finite_values)), abs(min(finite_values)), value_range, 1.0) * 0.012

    for bar, value in zip(bars, values):
        if value is None:
            continue

        y_position = value + offset if value >= 0 else value - offset
        vertical_alignment = "bottom" if value >= 0 else "top"

        axis.text(
            bar.get_x() + bar.get_width() / 2,
            y_position,
            f"{value:.2f}",
            ha="center",
            va=vertical_alignment,
            fontsize=7,
            rotation=90 if len(values) > 12 else 0,
        )


def _cell13_plot_performance(
    labels,
    metrics,
    values_by_metric,
    output_path,
    title,
    x_label,
):
    """Create one long figure with one bar-chart row per performance metric."""
    if not labels:
        print(f"[skip] No rows available for {output_path}")
        return

    if not metrics:
        print(f"[skip] No performance metrics available for {output_path}")
        return

    row_height = 2.15 if len(labels) <= 15 else 2.35
    figure_height = max(12, len(metrics) * row_height)

    fig, axes = plt.subplots(
        nrows=len(metrics),
        ncols=1,
        figsize=(20, figure_height),
        squeeze=False,
    )
    axes = axes[:, 0]

    positions = list(range(len(labels)))
    colors = [
        "tab:green" if (value or 0) >= 0 else "tab:red"
        for value in (
            values_by_metric[metrics[0]]
            if metrics
            else []
        )
    ]

    for axis, metric in zip(axes, metrics):
        values = values_by_metric[metric]
        plot_values = [value if value is not None else 0.0 for value in values]

        bars = axis.bar(
            positions,
            plot_values,
            color=colors,
            alpha=0.80,
            edgecolor="black",
            linewidth=0.35,
        )

        axis.axhline(0, color="black", linewidth=0.8)
        axis.set_title(metric, loc="left", fontsize=11, fontweight="bold")
        axis.set_ylabel("Value")
        axis.set_xticks(positions)
        axis.set_xticklabels(labels, rotation=45, ha="right", fontsize=8)
        axis.grid(axis="y", alpha=0.30, linestyle="--")
        axis.set_axisbelow(True)

        _cell13_add_bar_labels(axis, bars, values)

        missing_positions = [
            position
            for position, value in enumerate(values)
            if value is None
        ]
        if missing_positions:
            axis.scatter(
                missing_positions,
                [0] * len(missing_positions),
                marker="x",
                color="gray",
                zorder=3,
                label="Missing value",
            )
            axis.legend(loc="best", fontsize=7)

    axes[-1].set_xlabel(x_label)
    fig.suptitle(title, fontsize=16, fontweight="bold", y=1.001)
    fig.tight_layout()

    output_path.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(output_path, dpi=160, bbox_inches="tight")
    plt.close(fig)
    print(f"Wrote {output_path}")


def create_performance_cycles_png(config):
    """Create one long multi-metric PNG for each performance cycle CSV."""
    if isinstance(config, str):
        config = json.loads(config)

    root = Path("../data/results") / config["expert"]
    performance_dir = root / "performance_cycles"

    for cycle_start in _cell13_cycle_dates(config):
        csv_path = performance_dir / f"performance_{cycle_start.isoformat()}.csv"
        if not csv_path.exists():
            print(f"[skip] Missing cycle performance CSV: {csv_path}")
            continue

        labels, metrics, values_by_metric = _cell13_read_performance_csv(
            csv_path,
            label_column="Pass",
        )

        _cell13_plot_performance(
            labels=labels,
            metrics=metrics,
            values_by_metric=values_by_metric,
            output_path=performance_dir / f"performance_{cycle_start.isoformat()}.png",
            title=(
                "Selected Configuration Performance — "
                f"Cycle Starting {cycle_start.isoformat()}"
            ),
            x_label="Selected configuration Pass",
        )


def create_performance_all_png(config):
    """Create one long multi-metric PNG comparing every cycle."""
    if isinstance(config, str):
        config = json.loads(config)

    root = Path("../data/results") / config["expert"]
    csv_path = root / "performance_all.csv"

    if not csv_path.exists():
        print(f"[skip] Missing overall performance CSV: {csv_path}")
        return

    labels, metrics, values_by_metric = _cell13_read_performance_csv(
        csv_path,
        label_column="cycle_FromDate",
    )

    _cell13_plot_performance(
        labels=labels,
        metrics=metrics,
        values_by_metric=values_by_metric,
        output_path=root / "performance_all.png",
        title="Performance Across All Optimization Cycles",
        x_label="Cycle start date",
    )


def create_performance_png_reports(config):
    if isinstance(config, str):
        config = json.loads(config)

    create_performance_cycles_png(config)
    create_performance_all_png(config)


create_performance_png_reports(CONFIG)
```
