```python
# Cell_05 create_cycle_folders
import json
from datetime import date, timedelta
from pathlib import Path


def _format_ini_date(d):
    """MT5 .ini wants YYYY.MM.DD format."""
    return d.strftime("%Y.%m.%d")


def _write_ini_from_dict(tester_dict, dest_path, symbol, from_date, to_date):
    """
    Write an MT5-style .ini file from a dict shaped like the JSON example.
    tester_dict must contain "Tester" and "TesterInputs" keys.
    """
    tester = dict(tester_dict["Tester"])
    inputs = dict(tester_dict["TesterInputs"])

    # Override the dynamic fields
    tester["Symbol"] = symbol
    tester["FromDate"] = _format_ini_date(from_date)
    tester["ToDate"] = _format_ini_date(to_date)

    lines = ["[Tester]"]
    for key, value in tester.items():
        lines.append(f"{key}={value}")

    lines.append("[TesterInputs]")
    for key, value in inputs.items():
        lines.append(f"{key}={value}")

    dest_path.write_text("\n".join(lines) + "\n", encoding="utf-8")


def _split_lookback(lookback, splits):
    """
    Split `lookback` days into `splits` approximately equal integer segments.
    Returns a list of segment lengths (ints) summing exactly to `lookback`.
    """
    if splits <= 0:
        raise ValueError("optimization_lookback_splits must be positive")
    base = lookback // splits
    remainder = lookback % splits
    # Distribute the remainder over the first `remainder` segments
    return [base + (1 if i < remainder else 0) for i in range(splits)]


def create_cycle_folders(config, set_1, set_2):
    """Create one folder per optimization cycle and return the cycle count.

    For each cycle folder, also write <set_name>.ini files derived from the
    given set definitions (dicts matching the JSON schema).

    In addition to set_1.ini and set_2.ini, this writes split variants
    set_2_<split>.ini (2-digit split index) where the optimization lookback
    window is divided into `optimization_lookback_splits` approximately equal
    segments.
    """
    if isinstance(config, str):
        config = json.loads(config)

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    lookback = int(config["optimization_lookback_days"])
    splits = int(config["optimization_lookback_splits"])

    if interval <= 0:
        raise ValueError("optimization_interval_days must be positive")
    if lookback < 0:
        raise ValueError("optimization_lookback_days must be non-negative")
    if splits <= 0:
        raise ValueError("optimization_lookback_splits must be positive")
    if end < start:
        raise ValueError("end_date must be on or after start_date")

    days = (end - start).days
    cycles = (days + interval - 1) // interval  # ceil division

    folder_name = f"{config['expert']}"
    expert_dir = Path("../data/results") / folder_name
    expert_dir.mkdir(parents=True, exist_ok=True)

    # Explicit (name, dict) pairs — no reliance on a 'name' key inside the dict.
    sets = [
        ("set_1", set_1),  # ("EA_NBO_WeightedPricePDF_1", set_1),
        ("set_2", set_2),  # ("EA_NBO_WeightedPricePDF_2", set_2),
    ]

    # Precompute lookback segment lengths for the split set_2 files.
    lookback_segments = _split_lookback(lookback, splits)

    for cycle in range(0, cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        from_date = cycle_start - timedelta(days=lookback)

        cycle_dir = expert_dir / f"{config['symbol']}_{cycle_start.isoformat()}"
        cycle_dir.mkdir(parents=True, exist_ok=True)

        # Write the original set_1.ini and set_2.ini (unchanged behavior).
        for set_name, set_def in sets:
            _write_ini_from_dict(
                tester_dict=set_def,
                dest_path=cycle_dir / f"{set_name}.ini",
                symbol=config["symbol"],
                from_date=from_date,
                to_date=cycle_start,
            )

        # Write split variants of set_2: set_2_<split>.ini (2-digit index).
        # The lookback window [from_date, cycle_start) is divided into
        # consecutive, approximately equal segments.
        split_from = from_date
        for split_idx, seg_len in enumerate(lookback_segments):
            split_to = split_from + timedelta(days=seg_len)
            _write_ini_from_dict(
                tester_dict=set_2,
                dest_path=cycle_dir / f"set_2_{split_idx:02d}.ini",
                symbol=config["symbol"],
                from_date=split_from,
                to_date=split_to,
            )
            split_from = split_to

    return cycles


cycle_count = create_cycle_folders(
    CONFIG, EA_NBO_WeightedPricePDF_1, EA_NBO_WeightedPricePDF_2
)
print(f"Created {cycle_count} cycle folders.")
```

```python
# Cell_06 run_cycle_backtests
def run_cycle_backtests(config, timeout=3600, split_timeout=1800):
    """
    For each cycle folder:
      1. Run set_1.ini (ignore its report).
      2. Run set_2.ini, save its report as set_2.xml.
      3. Run every set_2_<split>.ini (2-digit split) next to set_2.ini with the same
         logic, saving its report as set_2_<split>.xml.
         Each split's INI Report= path is rewritten to a unique file so splits
         don't collide with set_2.ini (and with each other).

    Original INI files are restored after execution.
    """
    import shutil
    import subprocess
    import time
    import re
    from datetime import date, timedelta
    from pathlib import Path

    terminal_path = config["terminal_path"]
    mql5_path = Path(config["mql5_path"])
    report_path = Path(config["reprort_path"])

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    # Mirror create_cycle_folders logic to enumerate cycles deterministically.
    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _wait_for_stable_file(path, timeout=timeout, poll=1.0, stable_for=2.0):
        deadline = time.time() + timeout
        last_size = -1
        stable_since = None
        while time.time() < deadline:
            if path.exists():
                try:
                    size = path.stat().st_size
                except OSError:
                    size = -1
                if size > 0:
                    if size == last_size:
                        if stable_since is None:
                            stable_since = time.time()
                        elif time.time() - stable_since >= stable_for:
                            return True
                    else:
                        stable_since = None
                        last_size = size
            time.sleep(poll)
        return False

    def _resolve_report_path(ini_tester):
        report = ini_tester.get("Report", "")
        if not report:
            return None
        rp = Path(report)
        if rp.is_absolute():
            return rp
        candidate = report_path / rp.name
        if candidate.exists():
            return candidate
        return mql5_path / rp

    def _run_ini(ini_path):
        return subprocess.Popen([terminal_path, f"/config:{ini_path}"])

    def _parse_tester(ini_path):
        try:
            import configparser
            parser = configparser.ConfigParser()
            parser.read(ini_path, encoding="utf-8")
            return dict(parser["Tester"]) if parser.has_section("Tester") else {}
        except Exception as exc:
            print(f"[warn] Could not parse {ini_path}: {exc}")
            return {}

    def _run_and_capture(ini_path, report_file, label, timeout=timeout):
        if report_file.exists():
            try:
                report_file.unlink()
            except OSError:
                pass

        print(f"[cycle] {label}: running {ini_path.name} (report -> {report_file})")
        proc = _run_ini(ini_path)
        try:
            proc.wait(timeout=timeout)
        except subprocess.TimeoutExpired:
            print(f"[warn] {label} timed out; terminating.")
            proc.terminate()
            try:
                proc.wait(timeout=30)
            except subprocess.TimeoutExpired:
                proc.kill()

        if not _wait_for_stable_file(report_file, timeout=timeout):
            print(f"[warn] {label} report not produced: {report_file}")
            return False
        return True

    def _rewrite_report_path(ini_path, new_report_value):
        """
        Rewrite the `Report=` line under [Tester] in `ini_path` in place.
        Returns the original bytes so the caller can restore afterwards.
        """
        original = ini_path.read_bytes()
        text = original.decode("utf-8", errors="replace")
        lines = text.splitlines(keepends=True)
        in_tester = False
        out_lines = []
        for line in lines:
            stripped = line.strip()
            if stripped.startswith("[") and stripped.endswith("]"):
                in_tester = (stripped.lower() == "[tester]")
                out_lines.append(line)
                continue
            if in_tester and re.match(r"\s*Report\s*=", line, flags=re.IGNORECASE):
                indent = line[: len(line) - len(line.lstrip())]
                eol = "\r\n" if line.endswith("\r\n") else ("\n" if line.endswith("\n") else "")
                out_lines.append(f"{indent}Report={new_report_value}{eol}")
            else:
                out_lines.append(line)
        ini_path.write_bytes("".join(out_lines).encode("utf-8"))
        return original

    for cycle_dir in cycle_dirs:
        set_1_ini = cycle_dir / "set_1.ini"
        set_2_ini = cycle_dir / "set_2.ini"

        if not set_1_ini.exists() or not set_2_ini.exists():
            print(f"[skip] Missing INI(s) in {cycle_dir}")
            continue

        # Discover additional set_2_<split>.ini files (2-digit split).
        extra_set_2_inis = []
        for candidate in sorted(cycle_dir.iterdir()):
            if candidate.is_file() and candidate != set_2_ini:
                m = re.fullmatch(r"set_2_(\d{2})\.ini", candidate.name)
                if m:
                    extra_set_2_inis.append((m.group(1), candidate))

        # Backups: set_1, set_2, and each extra split.
        ini_backups = {}
        for ini_path in [set_1_ini, set_2_ini] + [p for _, p in extra_set_2_inis]:
            try:
                ini_backups[ini_path] = ini_path.read_bytes()
            except OSError as exc:
                print(f"[warn] Could not back up {ini_path}: {exc}")

        set_2_tester = _parse_tester(set_2_ini)
        set_2_report = _resolve_report_path(set_2_tester) or (report_path / "Report_2.xml")

        try:
            # --- Step 1: set_1 (ignored) ---
            if set_2_report.exists():
                try:
                    set_2_report.unlink()
                except OSError:
                    pass
            print(f"[cycle] {cycle_dir.name}: running set_1.ini")
            proc1 = _run_ini(set_1_ini)
            try:
                proc1.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] set_1 timed out; terminating.")
                proc1.terminate()
                try:
                    proc1.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc1.kill()
            time.sleep(3)

            # --- Step 2: set_2 (report kept as set_2.xml) ---
            if _run_and_capture(set_2_ini, set_2_report,
                                f"{cycle_dir.name}:set_2", timeout=timeout):
                dest_report = cycle_dir / "set_2.xml"
                shutil.copy2(set_2_report, dest_report)
                print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                      f"({dest_report.stat().st_size} bytes)")

            # --- Step 3: extra splits (unique report path per split) ---
            for split, extra_ini in extra_set_2_inis:
                # Give each split its own report file so it can't collide
                # with set_2.ini's Report_2.xml (the cause of the hang).
                split_report_name = f"Report_2_{split}.xml"
                extra_report = report_path / split_report_name

                # Rewrite the split's INI Report= to the unique file.
                try:
                    _rewrite_report_path(
                        extra_ini, f"Reports\\AdaptiveTrader\\{split_report_name}"
                    )
                except OSError as exc:
                    print(f"[warn] Could not rewrite Report= in {extra_ini}: {exc}")
                    continue

                label = f"{cycle_dir.name}:set_2_{split}"
                try:
                    ok = _run_and_capture(
                        extra_ini, extra_report, label, timeout=split_timeout
                    )
                    if ok:
                        dest_report = cycle_dir / f"set_2_{split}.xml"
                        shutil.copy2(extra_report, dest_report)
                        print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                              f"({dest_report.stat().st_size} bytes)")
                finally:
                    # Restore this split's original INI immediately.
                    try:
                        extra_ini.write_bytes(ini_backups[extra_ini])
                    except OSError as exc:
                        print(f"[warn] Failed to restore {extra_ini}: {exc}")

        finally:
            # Restore set_1 / set_2 (splits were already restored above).
            for ini_path in (set_1_ini, set_2_ini):
                backup = ini_backups.get(ini_path)
                if backup is not None:
                    try:
                        ini_path.write_bytes(backup)
                    except OSError as exc:
                        print(f"[warn] Failed to restore {ini_path}: {exc}")
            time.sleep(3)

    print("All cycles completed.")


run_cycle_backtests(CONFIG)
```

```python
# Cell_10 select_top_configs_splits
def select_top_configs_splits(config):
    """
    For each cycle folder that contains set_2_<split>.xml files (2-digit split)
    next to set_2.xml, build a DataFrame of the form:

        [evaluation_columns] + [parameters_columns] + [Result_<split>...]

    where:
      - rows come from set_2.xml (only rows with Result >= 40)
      - Result_<split> for each row is the Result of the matching row
        (matched on parameters_columns) in set_2_<split>.xml, or 0 if missing.
      - "Result" = product(Result_<split>) / 100^(n_splits - 1), rounded to 2 dp.
      - evaluation columns (except Pass, Result) = average of the corresponding
        values from all set_2_<split>.xml files, ignoring files where the row
        is absent.
      - rows are sorted by Result descending; Pass renumbered 1..N.
      - Merge rows sharing the same evaluation columns (except Pass).
      - condition columns (O01..O07, C01..C07) keep the shared value only if
        all rows agree, otherwise 0; other columns keep the most-repeated
        value (ties broken by first occurrence).
      - top N = config["top_configs"] rows are selected.
      - Result_<split> columns are dropped, and the final DataFrame is written
        to "selected_configs_splits.csv" in the cycle folder, with "Pass"
        renumbered 1..N, preserving the original column order of the header.
    """
    import csv
    import re
    import xml.etree.ElementTree as ET
    from collections import Counter
    from pathlib import Path
    from datetime import date, timedelta
    import pandas as pd

    evaluation_columns = [
        "Pass", "Result", "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]
    condition_columns = [
        "O01", "O02", "O03", "O04", "O05", "O06", "O07",
        "C01", "C02", "C03", "C04", "C05", "C06", "C07",
    ]

    top_n = int(config["top_configs"])
    if top_n <= 0:
        raise ValueError("top_configs must be positive")

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _strip_namespaces(xml_text):
        xml_text = re.sub(r'\sxmlns(:\w+)?="[^"]*"', "", xml_text)
        xml_text = re.sub(r'\s\w+:\w+="[^"]*"', "", xml_text)
        xml_text = re.sub(r"<(/?)\w+:", r"<\1", xml_text)
        xml_text = re.sub(r"<\?mso-application[^?]*\?>", "", xml_text)
        return xml_text

    def _parse_report(xml_path):
        """Return (ordered_column_names, list_of_row_values_lists)."""
        raw = xml_path.read_text(encoding="utf-8", errors="replace")
        raw = _strip_namespaces(raw)
        root = ET.fromstring(raw)

        rows = root.findall(".//Row")
        if not rows:
            rows = [el for el in root.iter() if el.tag.endswith("Row")]
        if not rows:
            raise ValueError(f"No Row elements found in {xml_path}")

        def _row_values(row_el):
            values = []
            for cell in row_el:
                if not cell.tag.endswith("Cell"):
                    continue
                data_el = None
                for child in cell:
                    if child.tag.endswith("Data"):
                        data_el = child
                        break
                if data_el is not None:
                    values.append(data_el.text if data_el.text is not None else "")
                else:
                    values.append(cell.text if cell.text is not None else "")
            return values

        header = _row_values(rows[0])

        result_rows = []
        for row_el in rows[1:]:
            vals = _row_values(row_el)
            if not vals or vals[0] == "":
                continue
            try:
                int(vals[0])
            except (TypeError, ValueError):
                continue
            result_rows.append(vals)

        return header, result_rows

    def _to_float(v):
        try:
            return float(v)
        except (TypeError, ValueError):
            return None

    def _most_repeated(values):
        if not values:
            return ""
        counts = Counter(values)
        max_count = max(counts.values())
        for v in values:
            if counts[v] == max_count:
                return v
        return values[0]

    def _load_split_reports(cycle_dir):
        """Return dict: split_str -> (header, rows) for each set_2_<split>.xml."""
        splits = {}
        for candidate in sorted(cycle_dir.iterdir()):
            if candidate.is_file():
                m = re.fullmatch(r"set_2_(\d{2})\.xml", candidate.name)
                if m:
                    try:
                        header, rows = _parse_report(candidate)
                    except Exception as exc:
                        print(f"[warn] Failed to parse {candidate}: {exc}")
                        continue
                    splits[m.group(1)] = (header, rows)
        return splits

    def _build_param_index(header, rows):
        """Return dict: tuple(param_values) -> row_value_list."""
        param_idx = [i for i, c in enumerate(header) if c not in evaluation_columns]
        index = {}
        for row in rows:
            key = tuple(row[i] if i < len(row) else "" for i in param_idx)
            index[key] = row
        return param_idx, index

    def _build_base_dataframe(base_header, base_rows):
        """Build dataframe from set_2.xml, only rows with Result >= 40."""
        try:
            result_idx = base_header.index("Result")
        except ValueError:
            raise ValueError("'Result' column not found in set_2.xml header")

        for col in evaluation_columns:
            if col not in base_header:
                raise ValueError(f"Missing evaluation column: {col}")

        param_idx = [i for i, c in enumerate(base_header) if c not in evaluation_columns]
        param_cols = [base_header[i] for i in param_idx]

        # Order: evaluation + parameters (preserve header order)
        ordered_cols = [c for c in base_header]

        records = []
        for row in base_rows:
            if result_idx >= len(row):
                continue
            res = _to_float(row[result_idx])
            if res is None or res < 40:
                continue
            records.append({c: (row[i] if i < len(row) else "")
                            for i, c in enumerate(base_header)})

        if not records:
            return pd.DataFrame(columns=ordered_cols), param_cols

        df = pd.DataFrame(records, columns=ordered_cols)
        return df, param_cols

    def _augment_with_splits(df, param_cols, base_header, split_reports):
        """Add Result_<split> columns and average the evaluation columns."""
        # Build per-split lookup on parameters only.
        split_lookups = {}
        for split, (sheader, srows) in split_reports.items():
            s_param_idx = [i for i, c in enumerate(sheader) if c not in evaluation_columns]
            s_param_cols = [sheader[i] for i in s_param_idx]
            s_result_idx = sheader.index("Result") if "Result" in sheader else None

            lookup = {}
            for row in srows:
                key = tuple(row[i] if i < len(row) else "" for i in s_param_idx)
                lookup[key] = row
            split_lookups[split] = {
                "param_cols": s_param_cols,
                "result_idx": s_result_idx,
                "header": sheader,
                "lookup": lookup,
            }

        # Make sure parameter columns are usable for key construction.
        # Some params in base might not exist in split; fall back to "".
        for split, info in split_lookups.items():
            missing = [c for c in info["param_cols"] if c not in df.columns]
            for c in missing:
                df[c] = ""
            if info["result_idx"] is None:
                raise ValueError(f"'Result' column missing in split {split}")

        # Fill Result_<split> columns.
        for split, info in split_lookups.items():
            col_name = f"Result_{split}"
            values = []
            for _, row in df.iterrows():
                key = tuple(str(row.get(c, "")) for c in info["param_cols"])
                matched = info["lookup"].get(key)
                if matched is None:
                    values.append(0.0)
                else:
                    r = _to_float(matched[info["result_idx"]]
                                  if info["result_idx"] < len(matched) else "")
                    values.append(r if r is not None else 0.0)
            df[col_name] = values

        # Compute Result from product of Result_<split>.
        split_cols = [f"Result_{s}" for s in split_lookups.keys()]
        n_splits = len(split_cols)
        if n_splits > 0:
            product = df[split_cols[0]].astype(float)
            for c in split_cols[1:]:
                product = product * df[c].astype(float)
            df["Result"] = (product / (100.0 ** (n_splits - 1))).round(2)

        # Average evaluation columns (except Pass, Result) across splits.
        avg_eval_cols = [c for c in evaluation_columns if c not in ("Pass", "Result")]
        for split, info in split_lookups.items():
            sheader = info["header"]
            for c in avg_eval_cols:
                if c not in sheader:
                    continue
                s_idx = sheader.index(c)

                def _avg_for_row(row):
                    key = tuple(str(row.get(pc, "")) for pc in info["param_cols"])
                    matched = info["lookup"].get(key)
                    if matched is None or s_idx >= len(matched):
                        return None
                    v = _to_float(matched[s_idx])
                    return v

                # collect values
                collected = df.apply(_avg_for_row, axis=1)

                if c in df.columns:
                    # keep running (sum, count)
                    df[c] = df[c].astype(object)
                    # We'll do a two-pass aggregation instead.
                else:
                    df[c] = None

                # Use separate accumulation for clarity.
                # Store per-split values in temp dict keyed by index.
                if "_split_eval_acc" not in df.attrs:
                    df.attrs["_split_eval_acc"] = {}
                if c not in df.attrs["_split_eval_acc"]:
                    df.attrs["_split_eval_acc"][c] = {"sum": {}, "count": {}}

                acc = df.attrs["_split_eval_acc"][c]
                for idx, val in collected.items():
                    if val is None:
                        continue
                    acc["sum"][idx] = acc["sum"].get(idx, 0.0) + float(val)
                    acc["count"][idx] = acc["count"].get(idx, 0) + 1

        # Finalize averages
        if "_split_eval_acc" in df.attrs:
            for c, acc in df.attrs["_split_eval_acc"].items():
                def _avg(idx):
                    n = acc["count"].get(idx, 0)
                    if n == 0:
                        return df.at[idx, c] if c in df.columns else None
                    return round(acc["sum"][idx] / n, 2)

                # Preserve original type for base-only cells; use averaged where possible.
                new_col = []
                for idx in df.index:
                    avg_v = _avg(idx)
                    if avg_v is None:
                        new_col.append(df.at[idx, c] if c in df.columns else "")
                    else:
                        new_col.append(avg_v)
                df[c] = new_col
            df.attrs.pop("_split_eval_acc", None)

        return df, split_lookups

    def _merge_rows(df, param_cols):
        """
        Merge rows sharing the same evaluation columns (except Pass).
        - Condition columns: shared value if all agree, else 0.
        - All other columns: most-repeated value (ties broken by first occurrence).
        """
        eval_keys = [c for c in evaluation_columns if c != "Pass"]
        # Restrict to columns that actually exist.
        group_keys = [c for c in eval_keys if c in df.columns]

        groups = {}
        for idx, row in df.iterrows():
            key = tuple(str(row.get(c, "")) for c in group_keys)
            groups.setdefault(key, []).append((idx, row))

        known = set(evaluation_columns) | set(condition_columns) | {"Result", "Pass"}
        merged_records = []

        for key, items in groups.items():
            rows = [r for _, r in items]
            first = rows[0]
            record = {}

            # Evaluation columns: keep first row's value (except Pass).
            for c in evaluation_columns:
                if c == "Pass":
                    continue
                record[c] = first.get(c, "")

            # Condition columns: shared or 0.
            for c in condition_columns:
                if c not in df.columns:
                    continue
                vals = [str(r.get(c, "")) for r in rows]
                uniq = set(vals)
                record[c] = vals[0] if len(uniq) == 1 else "0"

            # Other columns: most-repeated.
            for c in df.columns:
                if c in known:
                    continue
                vals = [str(r.get(c, "")) for r in rows]
                record[c] = _most_repeated(vals)

            merged_records.append(record)

        return merged_records

    def _write_splits_csv(merged_records, final_header, dest_path):
        with dest_path.open("w", encoding="utf-8", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=final_header, extrasaction="ignore")
            writer.writeheader()
            for rank, rec in enumerate(merged_records, start=1):
                row_out = dict(rec)
                row_out["Pass"] = rank
                writer.writerow(row_out)

    any_written = False
    for cycle_dir in cycle_dirs:
        base_report = cycle_dir / "set_2.xml"
        if not base_report.exists():
            continue

        split_reports = _load_split_reports(cycle_dir)
        if not split_reports:
            continue

        # 1. Parse base
        try:
            base_header, base_rows = _parse_report(base_report)
        except Exception as exc:
            print(f"[warn] Failed to parse {base_report}: {exc}")
            continue

        # 2. Build dataframe (Result >= 40)
        try:
            df, param_cols = _build_base_dataframe(base_header, base_rows)
        except Exception as exc:
            print(f"[warn] Failed to build base dataframe for {cycle_dir}: {exc}")
            continue

        if df.empty:
            print(f"[warn] No rows with Result >= 40 in {cycle_dir}")
            continue

        # 3. Add Result_<split> + averages
        try:
            df, split_lookups = _augment_with_splits(df, param_cols, base_header, split_reports)
        except Exception as exc:
            print(f"[warn] Failed to augment with splits for {cycle_dir}: {exc}")
            continue

        # 4. Sort by Result descending; renumber Pass 1..N (temporary).
        df = df.sort_values(by="Result", ascending=False, kind="stable").reset_index(drop=True)
        df["Pass"] = range(1, len(df) + 1)

        # 5. Merge on evaluation columns (except Pass).
        merged_records = _merge_rows(df, param_cols)
        merged_df = pd.DataFrame(merged_records)

        # 6. Sort merged by Result descending again.
        merged_df["_Result_sort"] = pd.to_numeric(merged_df["Result"], errors="coerce").fillna(0)
        merged_df = merged_df.sort_values("_Result_sort", ascending=False, kind="stable").reset_index(drop=True)

        # 7. Top N.
        selected = merged_df.head(top_n).copy()

        # 8. Final header: original order from base report header
        #    but drop Result_<split> columns.
        result_split_cols = [c for c in selected.columns if re.fullmatch(r"Result_\d{2}", c)]
        selected = selected.drop(columns=result_split_cols + ["_Result_sort"], errors="ignore")

        final_header = [c for c in base_header if c in selected.columns]
        # Ensure Pass is first and Result present.
        selected = selected[[c for c in final_header]]

        # Renumber Pass 1..N.
        selected["Pass"] = range(1, len(selected) + 1)

        dest_path = cycle_dir / "selected_configs_splits.csv"
        _write_splits_csv(selected.to_dict(orient="records"), final_header, dest_path)
        print(f"[cycle] {cycle_dir.name}: wrote {dest_path.name} "
              f"({len(selected)} of {len(merged_df)} merged configs, "
              f"{len(split_lookups)} splits)")
        any_written = True

    if not any_written:
        print("No cycle had split reports; nothing written.")
    else:
        print("Split-aware top configuration selection completed.")


select_top_configs_splits(CONFIG)
```

```python
# Cell_07 select_top_configs
def select_top_configs(config):
    """
    For each cycle folder:
      1. Read the "set_2.xml" report and extract the optimization results.
      2. Merge rows sharing the same "Result" value.
         - Evaluation columns keep the first row's value.
         - Condition columns keep the shared value only if all rows agree;
           otherwise 0 (per original spec).
         - All other columns keep the most-repeated value in the group
           (ties broken by first occurrence).
      3. Select the top N configurations by the optimization criterion.
      4. Write them to "set_2_selected.csv", preserving the original column
         order from the report header. The "Pass" column is renumbered 1..N.
    """
    import csv
    import re
    import xml.etree.ElementTree as ET
    from collections import Counter
    from pathlib import Path
    from datetime import date, timedelta

    evaluation_columns = [
        "Pass", "Result", "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]
    condition_columns = [
        "O01", "O02", "O03", "O04", "O05", "O06", "O07",
        "C01", "C02", "C03", "C04", "C05", "C06", "C07",
    ]

    top_n = int(config["top_configs"])
    if top_n <= 0:
        raise ValueError("top_configs must be positive")

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _strip_namespaces(xml_text):
        xml_text = re.sub(r'\sxmlns(:\w+)?="[^"]*"', "", xml_text)
        xml_text = re.sub(r'\s\w+:\w+="[^"]*"', "", xml_text)
        xml_text = re.sub(r"<(/?)\w+:", r"<\1", xml_text)
        xml_text = re.sub(r"<\?mso-application[^?]*\?>", "", xml_text)
        return xml_text

    def _parse_report(xml_path):
        """Return (ordered_column_names, list_of_row_values_lists)."""
        raw = xml_path.read_text(encoding="utf-8", errors="replace")
        raw = _strip_namespaces(raw)
        root = ET.fromstring(raw)

        rows = root.findall(".//Row")
        if not rows:
            rows = [el for el in root.iter() if el.tag.endswith("Row")]
        if not rows:
            raise ValueError(f"No Row elements found in {xml_path}")

        def _row_values(row_el):
            values = []
            for cell in row_el:
                if not cell.tag.endswith("Cell"):
                    continue
                data_el = None
                for child in cell:
                    if child.tag.endswith("Data"):
                        data_el = child
                        break
                if data_el is not None:
                    values.append(data_el.text if data_el.text is not None else "")
                else:
                    values.append(cell.text if cell.text is not None else "")
            return values

        header = _row_values(rows[0])

        result_rows = []
        for row_el in rows[1:]:
            vals = _row_values(row_el)
            if not vals or vals[0] == "":
                continue
            try:
                int(vals[0])
            except (TypeError, ValueError):
                continue
            result_rows.append(vals)

        return header, result_rows

    def _to_float(v):
        try:
            return float(v)
        except (TypeError, ValueError):
            return float("-inf")

    def _most_repeated(values):
        """Most common value; ties broken by first occurrence in the list."""
        if not values:
            return ""
        counts = Counter(values)
        max_count = max(counts.values())
        for v in values:  # preserve first-occurrence order
            if counts[v] == max_count:
                return v
        return values[0]

    def _merge_by_result(header, rows):
        try:
            result_idx = header.index("Result")
        except ValueError:
            raise ValueError("'Result' column not found in report header")

        for col in evaluation_columns:
            if col not in header:
                raise ValueError(f"Missing evaluation column: {col}")

        idx_by_name = {name: i for i, name in enumerate(header)}
        known = set(evaluation_columns) | set(condition_columns) | {"Result", "Pass"}

        groups = {}
        for row in rows:
            if result_idx >= len(row):
                continue
            groups.setdefault(row[result_idx], []).append(row)

        merged = []
        for key, group_rows in groups.items():
            merged_row = {}

            # Evaluation columns: first row's value.
            for col in evaluation_columns:
                i = idx_by_name[col]
                merged_row[col] = (
                    group_rows[0][i] if i < len(group_rows[0]) else ""
                )

            # Condition columns: shared value, else 0.
            for col in condition_columns:
                if col not in idx_by_name:
                    continue
                i = idx_by_name[col]
                values = [r[i] for r in group_rows if i < len(r)]
                uniq = set(values)
                merged_row[col] = uniq.pop() if len(uniq) == 1 else "0"

            # All other columns: most-repeated value in the group.
            for col, i in idx_by_name.items():
                if col in known:
                    continue
                values = [r[i] for r in group_rows if i < len(r)]
                merged_row[col] = _most_repeated(values)

            merged_row["Result"] = key
            merged.append(merged_row)

        return merged

    def _sort_value(merged_row):
        custom = _to_float(merged_row.get("Custom", ""))
        if custom != float("-inf"):
            return custom
        return _to_float(merged_row.get("Profit", ""))

    def _write_selected_csv(merged_rows, header, dest_path):
        with dest_path.open("w", encoding="utf-8", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=header, extrasaction="ignore")
            writer.writeheader()
            for rank, row in enumerate(merged_rows, start=1):
                row_out = dict(row)
                row_out["Pass"] = rank
                writer.writerow(row_out)

    for cycle_dir in cycle_dirs:
        report_path = cycle_dir / "set_2.xml"
        if not report_path.exists():
            print(f"[skip] Missing report in {cycle_dir}: {report_path}")
            continue

        try:
            header, rows = _parse_report(report_path)
        except Exception as exc:
            print(f"[warn] Failed to parse {report_path}: {exc}")
            continue

        try:
            merged = _merge_by_result(header, rows)
        except Exception as exc:
            print(f"[warn] Failed to merge results in {cycle_dir}: {exc}")
            continue

        if not merged:
            print(f"[warn] No usable result rows in {cycle_dir}")
            continue

        merged.sort(key=_sort_value, reverse=True)
        selected = merged[:top_n]

        dest_path = cycle_dir / "selected_configs.csv"
        _write_selected_csv(selected, header, dest_path)
        print(f"[cycle] {cycle_dir.name}: wrote {dest_path.name} "
              f"({len(selected)} of {len(merged)} merged configs)")

    print("Top configuration selection completed.")


select_top_configs(CONFIG)

```

```python
# Cell_08 create_selected_inis
import csv
from datetime import date, timedelta
from pathlib import Path


def create_selected_inis(config):
    """
    For each cycle folder:
      1. Read the "selected_configs.csv".
      2. For each row, extract the parameters_columns columns.
      3. For each row, write a new INI file in the cycle folder named
         "selected_<Pass>.ini" (2-digit Pass) with the parameters filled in
         from the CSV row, using EA_NBO_WeightedPricePDF_3 as template.

    The Tester FromDate/ToDate are set to the cycle's start date and end date
    (cycle_start, cycle_end = cycle_start + interval_days).
    All "FILL_VALUE" placeholders in the TesterInputs section are replaced
    with values from the row.
    """
    evaluation_columns = [
        "Profit", "Expected Payoff", "Profit Factor",
        "Recovery Factor", "Sharpe Ratio", "Custom",
    ]

    # Load template by name (config["expert"] + "_3").
    template_name = f"{config['expert']}_3"
    template = globals().get(template_name)
    if template is None:
        raise ValueError(f"Template '{template_name}' not found in globals")

    if isinstance(config, str):
        import json
        config = json.loads(config)

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    def _format_ini_date(d):
        return d.strftime("%Y.%m.%d")

    def _fill_value(template_value, row, col_name):
        """Replace a 'FILL_VALUE||...' string with the row's value for col_name."""
        if not isinstance(template_value, str):
            return template_value
        if not template_value.startswith("FILL_VALUE"):
            return template_value
        # Strip the leading FILL_VALUE placeholder and the leading '||'.
        rest = template_value[len("FILL_VALUE"):]
        if rest.startswith("||"):
            rest = rest[2:]
        # Replace only the first segment (the actual value), keep the rest.
        value = row.get(col_name, "")
        return f"{value}||{rest}" if rest else str(value)

    def _write_ini(template_dict, row, dest_path, from_date, to_date):
        tester = dict(template_dict["Tester"])
        inputs = dict(template_dict["TesterInputs"])

        tester["Symbol"] = config["symbol"]
        tester["FromDate"] = _format_ini_date(from_date)
        tester["ToDate"] = _format_ini_date(to_date)

        for key, val in inputs.items():
            inputs[key] = _fill_value(val, row, key)

        lines = ["[Tester]"]
        for k, v in tester.items():
            lines.append(f"{k}={v}")
        lines.append("[TesterInputs]")
        for k, v in inputs.items():
            lines.append(f"{k}={v}")
        dest_path.write_text("\n".join(lines) + "\n", encoding="utf-8")

    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_end = cycle_start + timedelta(days=interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if not cycle_dir.exists():
            print(f"[skip] Missing cycle folder: {cycle_dir}")
            continue

        csv_path = cycle_dir / "selected_configs.csv"
        if not csv_path.exists():
            print(f"[skip] Missing CSV: {csv_path}")
            continue

        with csv_path.open("r", encoding="utf-8", newline="") as f:
            reader = csv.DictReader(f)
            rows = list(reader)

        if not rows:
            print(f"[warn] Empty CSV: {csv_path}")
            continue

        # Validate evaluation columns exist in the CSV header.
        fieldnames = reader.fieldnames or []
        for col in evaluation_columns:
            if col not in fieldnames:
                raise ValueError(f"Missing evaluation column '{col}' in {csv_path}")

        for row in rows:
            try:
                pass_num = int(row["Pass"])
            except (KeyError, ValueError, TypeError):
                print(f"[warn] Bad Pass value in {csv_path}: {row.get('Pass')!r}")
                continue

            dest_path = cycle_dir / f"selected_{pass_num:02d}.ini"
            _write_ini(
                template_dict=template,
                row=row,
                dest_path=dest_path,
                from_date=cycle_start,
                to_date=cycle_end,
            )

        print(f"[cycle] {cycle_dir.name}: wrote {len(rows)} selected INI(s)")

    print("Selected INI generation completed.")


create_selected_inis(CONFIG)
```

```python
# Cell_09 run_selected_backtests
def run_selected_backtests(config, timeout=3600):
    """
    For each cycle folder:
      1. look for all INI file in the cycle folder named "selected_<Pass>.ini" (2-digit Pass) with the parameters filled in
      2. run each INI file and wait for its MT5 report to be generated. wait for its report.
      3. Copy the report to the cycle folder as "selected_<Pass>.htm".

    Build the cycle list (mirrors create_cycle_folders logic).
    """
    import re
    import shutil
    import subprocess
    import time
    import configparser
    from datetime import date, timedelta
    from pathlib import Path

    terminal_path = config["terminal_path"]
    mql5_path = Path(config["mql5_path"])
    report_path = Path(config["reprort_path"])

    cycles_folder = Path("../data/results") / f"{config['expert']}"
    if not cycles_folder.exists():
        raise FileNotFoundError(f"Cycles folder not found: {cycles_folder}")

    start = date.fromisoformat(config["start_date"])
    end = date.fromisoformat(config["end_date"])
    interval = int(config["optimization_interval_days"])
    days = (end - start).days
    cycles = (days + interval - 1) // interval

    cycle_dirs = []
    for cycle in range(cycles):
        cycle_start = start + timedelta(days=cycle * interval)
        cycle_dir = cycles_folder / f"{config['symbol']}_{cycle_start.isoformat()}"
        if cycle_dir.exists():
            cycle_dirs.append(cycle_dir)

    def _wait_for_stable_file(path, timeout=timeout, poll=1.0, stable_for=2.0):
        """Wait until `path` exists, is non-empty, and its size has been stable."""
        deadline = time.time() + timeout
        last_size = -1
        stable_since = None
        while time.time() < deadline:
            if path.exists():
                try:
                    size = path.stat().st_size
                except OSError:
                    size = -1
                if size > 0:
                    if size == last_size:
                        if stable_since is None:
                            stable_since = time.time()
                        elif time.time() - stable_since >= stable_for:
                            return True
                    else:
                        stable_since = None
                        last_size = size
            time.sleep(poll)
        return False

    def _resolve_report_path(ini_tester):
        """Resolve the report path from the INI's Tester section."""
        report = ini_tester.get("Report", "")
        if not report:
            return None
        rp = Path(report)
        if rp.is_absolute():
            return rp
        candidate = report_path / rp.name
        if candidate.exists():
            return candidate
        return mql5_path / rp

    def _run_ini(ini_path):
        cmd = [terminal_path, f"/config:{ini_path}"]
        return subprocess.Popen(cmd)

    for cycle_dir in cycle_dirs:
        selected_inis = sorted(cycle_dir.glob("selected_*.ini"))
        if not selected_inis:
            print(f"[skip] No selected INIs in {cycle_dir}")
            continue

        for ini_path in selected_inis:
            match = re.fullmatch(r"selected_(\d+)\.ini", ini_path.name)
            if not match:
                continue
            pass_str = match.group(1)
            dest_report = cycle_dir / f"selected_{pass_str}.htm"

            # Parse Tester section to find the report path.
            try:
                parser = configparser.ConfigParser()
                parser.read(ini_path, encoding="utf-8")
                tester = dict(parser["Tester"]) if parser.has_section("Tester") else {}
            except Exception as exc:
                print(f"[warn] Could not parse {ini_path}: {exc}")
                tester = {}

            mt5_report = _resolve_report_path(tester) or (report_path / f"Report_3.htm")

            print(f"[cycle] {cycle_dir.name}: running {ini_path.name}")
            if mt5_report.exists():
                try:
                    mt5_report.unlink()
                except OSError:
                    pass

            proc = _run_ini(ini_path)
            try:
                proc.wait(timeout=timeout)
            except subprocess.TimeoutExpired:
                print(f"[warn] {ini_path.name} timed out for {cycle_dir.name}; terminating.")
                proc.terminate()
                try:
                    proc.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    proc.kill()

            if not _wait_for_stable_file(mt5_report, timeout=timeout):
                print(f"[warn] Report not produced for {ini_path.name} in {cycle_dir.name}: {mt5_report}")
                time.sleep(3)
                continue

            shutil.copy2(mt5_report, dest_report)
            print(f"[cycle] {cycle_dir.name}: saved {dest_report.name} "
                  f"({dest_report.stat().st_size} bytes)")

            # Allow MT5 to finish its shutdown before the next run.
            time.sleep(3)

    print("All selected backtests completed.")


run_selected_backtests(CONFIG)
```
